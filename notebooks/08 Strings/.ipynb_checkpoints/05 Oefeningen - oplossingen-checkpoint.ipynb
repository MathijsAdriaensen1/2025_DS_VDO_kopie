#### Vraag 1

Gegeven deze string van door komma's gescheiden karakters: maak drie nieuwe variabelen aan die de unicode-codepunten (in hexadecimaal), de hoofdletter en de kleine letterversies van elk karakter bevatten (ook door komma's gescheiden).
Bijvoorbeeld, als de string `'a, b, c'` was dan genereer je de volgende drie lijsten:
* `['0x61', '0x62', '0x63']`
* `['a', 'b', 'c']`
* `['A', 'B', 'C']`

[Je zal onder meer de`split()` en `strip()` functies moeten gebruiken...]

````python
# oplossing
unicode_lijst = []
hoofdletter_lijst = []
kleineletter_lijst = []

s = 'Π, ύ, θ, ω, ν'

for char in s.split(','):
    unicode_lijst.append(hex(ord(char.strip())))
    hoofdletter_lijst.append(char.strip().upper())
    kleineletter_lijst.append(char.strip().lower())

print(unicode_lijst)
print(hoofdletter_lijst)
print(kleineletter_lijst)
````

#### Vraag 2

Gebruikmakend van twee soorten stringinterpolatie, en gegeven de variabele `a` die een integer bevat, druk de volgende string af voor `a`:
`Het getal ... is even (of oneven)`

Bijvoorbeeld voor `a` gelijk aan '20' druk je af: `Het getal 20 is even`

Je krijgt twee variabelen `a` en `b` (met `b` niet gelijk aan nul), en je moet een string genereren als volgt:
```
'a / b = (result)'
```

````python
# Gegeven variabele a
a = 20

# Oplossing met f-strings
even_of_oneven = "even" if a % 2 == 0 else "oneven"
print(f"Het getal {a} is {even_of_oneven}")

# Oplossing met de format() methode
print("Het getal {} is {}".format(a, even_of_oneven))
````

#### Vraag 3

Je wilt dat je string mooi wordt opgemaakt voor weergavedoeleinden, dus je wilt het tonen van mogelijke cijfers na de komma beperken in al je waarden tot 4 cijfers.

````python
# Gegeven variabelen a en b
a = 20
b = 3

# Bereken het resultaat van de deling
result = a / b

# Oplossing met f-strings, afronding tot 4 decimalen
print(f"{a} / {b} = {result:.4f}")

# Oplossing met de format() methode, afronding tot 4 decimalen
print("{:.0f} / {:.0f} = {:.4f}".format(a, b, result))
````

#### Vraag 4

Tel hoeveel er van iedere klinker (a, e, i, o, u) staan in een tekst string,en druk die teller af voor iedere klinker met een enkele geformatteerde string. Bedenk dat klinkers zowel hoofd- als kleine letters kunnen zijn."

````python
s="Tel hoeveel er van iedere klinker (a, e, i, o, u) staan in een tekst string,en druk die teller af voor iedere klinker met een enkele geformatteerde string. Bedenk dat klinkers zowel hoofd- als kleine letters kunnen zijn."
````

````python
klinkers = list('aeiou')

aantallen = {}
def vind(s, klinker, start_index=0):    
    start = s.find(klinker, start_index)
    if 0 <= start:
        if aantallen.get(klinker,0):
            aantallen[klinker] += 1
        else: 
            aantallen[klinker] = 1
        vind(s,klinker, start_index=start+1)
        
for klinker in klinkers:
    vind(s, klinker)

zin = "Ik tel " 
for klinker, aantal in aantallen.items():
    zin += f"{aantal} keer {klinker}, "
zin[:-2]
````

#### Vraag 5

Hieronder staat een tekst met een aantal tekens tussen vierkante haken.
Doorloop de tekst en druk alle tekens af die tussen vierkante haken staan.

````python
tekst = """ En ze stu[re]n [i] ngekleurde prentbriefkaarten van
plekken waarvan ze zich niet reali [s]eren dat ze er nooit
geweest zijn [a]an ' Iedereen op nummer 22, weer is prachti [g],
onz[e] kamer is aa[n] gekruisd . Missen jullie . E[t]en[ ]i[s]
vettig , maar we hebben een geweldig leuk restaurantje gevonden
in de achterstraatjes waar ze Heine [ke]n hebben en kaas en
uien chips en iemand die "Een beetje verliefd " speel [t] op een
a[c] cordeon ' en je zit vier dagen vast op Schip [h]ol voor je
vijfdaagse vliegvakantie met niks anders te eten dan
uitgedroogde voorverpakte boterhammen ... """
````

````python
def vinder(s, startteken, stopteken, start_index=0):    
    start = s.find(startteken, start_index)
    if 0 <= start:
        # zoek de stop
        stop = s.find(stopteken, start)
        print(s[start+1:stop])
        vinder(s,startteken, stopteken, start_index=start+1)
              
vinder(tekst, '[', ']')
````
