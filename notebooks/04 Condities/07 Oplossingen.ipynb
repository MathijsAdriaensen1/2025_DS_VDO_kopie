#### Oefening 1

Geef een variabele `a` (die een willekeurige waarde bevat), wijs de waarde `"N/A"` toe als `a` gelijk is aan `None`, en laat `a` anders ongewijzigd. Gebruik een `if...else...` instructie.

We moeten een variabele `a` creëren, deze een waarde toewijzen en vervolgens de waarde van `a` testen om te zien of we het moeten wijzigen naar `'N/A'`:

````python
a = 100

if a is None:
    a = 'N/A'
    
print(a)
````

Laten we nu onze code proberen met nog een paar mogelijke waarden voor `a`:

````python
a = 'Python'

if a is None:
    a = 'N/A'
    
print(a)
````

````python
a = None

if a is None:
    a = 'N/A'
    
print(a)
````

#### Oefening 2

Doe hetzelfde als Vraag 1, maar gebruik deze keer een ternaire operator.

Een ternaire operator berekent altijd een waarde, maar we willen `a` niet wijzigen als het niet gelijk is aan `None`.
To do this we'll simply have the ternary operator evaluate to `a` if `a` is not `None` and to `N/A` otherwise:

````python
a = 100
a = 'N/A' if a is None else a
print(a)
````

En opnieuw moeten we dit testen met een paar waarden:

````python
a = 'Python'
a = 'N/A' if a is None else a
print(a)
````

````python
a = None
a = 'N/A' if a is None else a
print(a)
````

#### Oefening 3

Gegeven een kredietscore `score`, wijs een stringwaarde toe aan een andere variabele `rating` op basis van de volgende schaal:
- [0, 580) --> Poor
- [580, 670) --> Fair
- [670, 740) --> Good
- [740, 800) --> Very Good
- [800, 850] --> Excellent


Dit kunnen we doen met behulp van een reeks `if...elif...else`-verklaringen:

````python
score = 720

if score < 580:
    rating = 'Poor'
elif score < 670:
    rating = 'Fair'
elif score < 740:
    rating = 'Good'
elif score < 800:
    rating = 'Very Good'
else:
    rating = 'Excellent'
    
print(rating)
````

Laten we dit testen met een paar waardes:

````python
score = 100
if score < 580:
    rating = 'Poor'
elif score < 670:
    rating = 'Fair'
elif score < 740:
    rating = 'Good'
elif score < 800:
    rating = 'Very Good'
else:
    rating = 'Excellent'
    
print(rating)
````

#### Oefening 4

Gegeven een `elapsed` tijd (in seconden), schrijf code om een variabele `magnitude` in te stellen op basis van de volgende condities:
- if elapsed time is less than 1 minute, `magnitude` --> `'seconds'`
- if elapsed time is more than 1 minute, but less than 1 hour, `magnitude` --> `'minutes'`
- if elapsed time is more than 1 hour, but less than 1 day, `magnitude` --> `'hours'`
- if elapsed time is more than 1 day, but less than 1 week: `magnitude` --> `'days'`
- if elapsed time is more than 1 week, `magnitude` --> '`weeks'`

Onze voorwaardelijke expressie zou er ongeveer zo uit kunnen zien (pseudo code = niet echte code):

```if elapsed < 1 minute:
    magnitude = 'seconds'
elif elapsed < 1 hour:
    magnitude = 'minutes'
 elif elapsed < 1 day:
     magnitude = 'hours'
 elif elapsed < 1 week:
     magnitude = 'days'
 else:
     magnitude = 'weeks'
```

Wat nu nog berekend moet worden, is wat `1 minuut`, `1 uur`, `1 dag` en `1 week` zijn in seconden (dat zijn de eenheden die gebruikt worden voor `elapsed`).

We berekenen die waarden en slaan ze op in variabelen - dit "decomposeert" niet alleen het probleem (breekt het in kleinere, meer beheersbare delen), maar zal ook helpen om de conditionele verklaring te verduidelijken zodat die erg veel lijkt op de pseudo-code hierboven.

````python
seconds_in_minute = 60
seconds_in_hour = 60 * seconds_in_minute
seconds_in_day = 24 * seconds_in_hour
seconds_in_week = 7 * seconds_in_day
````

We kunnen onze code nu op deze manier schrijven:

````python
elapsed = 23  # secs

if elapsed < seconds_in_minute:
    magnitude = 'seconds'
elif elapsed < seconds_in_hour:
    magnitude = 'minutes'
elif elapsed < seconds_in_day:
    magnitude = 'hours'
elif elapsed < seconds_in_week:
    magnitude = 'days'
else:
    magnitude = 'weeks'
    
print(magnitude)
````

Laten we dit proberen met een paar meer waarden:

````python
elapsed = 30 * 60  # 30 minutes in seconds

if elapsed < seconds_in_minute:
    magnitude = 'seconds'
elif elapsed < seconds_in_hour:
    magnitude = 'minutes'
elif elapsed < seconds_in_day:
    magnitude = 'hours'
elif elapsed < seconds_in_week:
    magnitude = 'days'
else:
    magnitude = 'weeks'
    
print(magnitude)
````

````python
elapsed = 12 * 60 * 60  # 12 hours in seconds

if elapsed < seconds_in_minute:
    magnitude = 'seconds'
elif elapsed < seconds_in_hour:
    magnitude = 'minutes'
elif elapsed < seconds_in_day:
    magnitude = 'hours'
elif elapsed < seconds_in_week:
    magnitude = 'days'
else:
    magnitude = 'weeks'
    
print(magnitude)
````
