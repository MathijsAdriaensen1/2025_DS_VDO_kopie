### *Intro:*

Er zijn 2 manieren om plots weer te geven: via de Pyplot State-Machine Interface ('Matlab Style') en via de object-georiënteerde API van Matplotlib.  

**De object-georiënteerde API van Matplotlib (fig, ax = plt.subplots() )**
De fig, ax = plt.subplots()-constructie wordt gebruikt in de object-georiënteerde benadering van Matplotlib, wat meer flexibiliteit en controle biedt, vooral wanneer je bijvoorbeeld meerdere subplots wilt maken binnen één figuur.  Met plt.subplots() kan je meerdere assenobjecten maken en individueel aanpassen.  

De object-georiënteerde benadering maakt het makkelijker om specifieke eigenschappen van figuren en assen aan te passen (bijvoorbeeld labels, titels, ticks, etc.).  

Als je meerdere figuren hebt en je wilt precies aangeven op welke figuur je een aanpassing wilt toepassen, kun je dat alleen doen met expliciete referenties naar fig en ax.

**De state-machine interface**  
De state-machine interface (plt.plot(), plt.xlabel(), enz.) maakt gebruik van een impliciete, globale figuur en assen. Deze interface houdt intern bij welke figuur en assen actief zijn, en alle aanroepen worden op de huidige, actieve assen toegepast. Dit maakt het gemakkelijk om snel en eenvoudig grafieken te maken, omdat je je geen zorgen hoeft te maken over het beheer van de objecten zelf.

Wanneer je in een Jupyter Notebook (of een andere interactieve omgeving) in verschillende cellen plt.plot() gebruikt, werkt Matplotlib standaard met een nieuwe figuur voor elke cel, tenzij je specifiek de actieve figuur behoudt. 

Wanneer je plt.show() aanroept in een cel, wordt de huidige figuur getoond en afgesloten. Dit betekent dat bij de volgende aanroep van plt.plot() een nieuwe figuur wordt gestart, tenzij je de figuur expliciet behoudt.

Wanneer je plt.plot() in een cel aanroept zonder dat er een actieve figuur is, maakt Matplotlib automatisch een nieuwe figuur aan. Dit is de standaard state-machine werking.

```python
# Solution version 1 - De object-georiënteerde API van Matplotlib

X = range(1, 50)
Y = [value * 3 for value in X]

# Print de waarden van X en Y
print("Values of X:")
print(*X) 
print("Values of Y (thrice of X):")
print(Y)

# Maak een figuur en assen aan
fig, ax = plt.subplots()

# Plot de lijn op de assen
ax.plot(X, Y)

# Labels en titel toevoegen
ax.set_xlabel('x - axis')
ax.set_ylabel('y - axis')
ax.set_title('Draw a line.')

# De grafiek weergeven
plt.show()
```

```python
# Solution

import datetime
from matplotlib import pyplot as plt
from matplotlib.dates import date2num

# date2num converteert een datum- of tijdobject (zoals datetime.datetime of datetime.date) naar een 
# float die het aantal dagen sinds een bepaald referentiepunt (in mpl 1 januari 0001) voorstelt. 
# Deze numerieke waarden zijn handig voor het plotten van tijdreeksen

data = [(datetime.datetime.strptime('2016-10-03', "%Y-%m-%d"), 772.559998),
        (datetime.datetime.strptime('2016-10-04', "%Y-%m-%d"), 776.429993),
        (datetime.datetime.strptime('2016-10-05', "%Y-%m-%d"), 776.469971),
        (datetime.datetime.strptime('2016-10-06', "%Y-%m-%d"), 776.859985),
        (datetime.datetime.strptime('2016-10-07', "%Y-%m-%d"), 775.080017 )]

x = [date2num(date) for (date, value) in data]
y = [value for (date, value) in data]

# Maak een figuur en een subplot aan
fig, ax = plt.subplots()

# Plot de data als een rode lijn met ronde markeringen
ax.plot(x, y, 'r-o')

# Zet de xtick-locaties
ax.set_xticks(x)

# Zet de xtick-labels
ax.set_xticklabels([date.strftime("%Y-%m-%d") for (date, value) in data])

# Toon de figuur
plt.show()
```

14. Write a Python program to display the grid and draw line charts of the closing value of Alphabet Inc. between October 3, 2016 to October 7, 2016. Customized the grid lines with rendering with a larger grid (major grid) and a smaller grid (minor grid). Turn on the grid but turn off ticks.
The code snippet gives the output shown in the following screenshot:

```python
img = mpimg.imread('../../IMAGES/matplotlib-basic-exercise-14.png')
plt.imshow(img)
plt.axis('off')  # Verberg de assen
plt.show()
```

```python
# Solution

data = [(datetime.datetime.strptime('2016-10-03', "%Y-%m-%d"), 772.559998),
        (datetime.datetime.strptime('2016-10-04', "%Y-%m-%d"), 776.429993),
        (datetime.datetime.strptime('2016-10-05', "%Y-%m-%d"), 776.469971),
        (datetime.datetime.strptime('2016-10-06', "%Y-%m-%d"), 776.859985),
        (datetime.datetime.strptime('2016-10-07', "%Y-%m-%d"), 775.080017)]

x = [date2num(date) for (date, value) in data]
y = [value for (date, value) in data]

# Maak een figuur en een subplot aan
fig, ax = plt.subplots()

# Plot the data as a red line with round markers
ax.plot(x,y,'r-o')

# Set the xtick locations
ax.set_xticks(x)

# Set the xtick labels
ax.set_xticklabels(
        [date.strftime("%Y-%m-%d") for (date, value) in data]
        )

# naming the x axis
plt.xlabel('Date')
# naming the y axis
plt.ylabel('Closing Value')
# giving a title  
plt.title('Closing stock value of Alphabet Inc.') 
# Turn on the minor TICKS, which are required for the minor GRID
plt.minorticks_on()

# Customize the major grid
plt.grid(which='major', linestyle='-', linewidth='0.5', color='red')
# Customize the minor grid
plt.grid(which='minor', linestyle=':', linewidth='0.5', color='black')

# Turn off the display of all ticks.
plt.tick_params(which='both', # Options for both major and minor ticks
                top='off', # turn off top ticks
                left='off', # turn off left ticks
                right='off',  # turn off right ticks
                bottom='off') # turn off bottom ticks
plt.show()
```

15. Write a  Python program to create multiple plots.

```python
img = mpimg.imread('../../IMAGES/matplotlib-basic-exercise-15.png')
plt.imshow(img)
plt.axis('off')  # Verberg de assen
plt.show()
```

```python
# Solution

# aanmaken figuur
fig = plt.figure()

# fig.subplots_adjust(): om de layout van de figuur aan te passen
# bottom=0.020: Stelt de ondermarge in op 2% van de hoogte van de figuur.
# left=0.020: Stelt de linkermarge in op 2% van de breedte van de figuur.
# top=0.900: Stelt de bovenzijde van de subplots in op 90% van de hoogte van de figuur.
# right=0.800: Stelt de rechterzijde van de subplots in op 80% van de breedte van de figuur.
fig.subplots_adjust(bottom=0.020, left=0.020, top = 0.900, right=0.800)

# plt.subplot(): om specifieke subplots te definiëren
# creëert een subplot in een raster van 2 rijen en 1 kolom, 
# en specificeert dat dit het eerste subplot is op de figuur
plt.subplot(2, 1, 1)

# Door op dezelfde figuur 2 rasters te definiëren waarvan het bovenste raster onderaan een rij heeft die
# niet wordt gebruikt, en het onderste raster bovenaan een rij heeft die niet wordt gebruikt,
# lopen deze niet visueel in elkaar over.

# plt.xticks(()) en plt.yticks(()) zorgen ervoor dat er 
# geen ticks en ticklabels worden weergegeven op respectievelijk de x-as en y-as van dit subplot.
plt.xticks(()), plt.yticks(())


# tweede subplot op dezelfde figuur; 2 rijen, 3 kolommen, figuur 4 dus eerste op 2de rij
plt.subplot(2, 3, 4) 
plt.xticks(())
plt.yticks(())

plt.subplot(2, 3, 5)# 2de op 2de rij
plt.xticks(())
plt.yticks(())

plt.subplot(2, 3, 6) # 3de op 2de rij
plt.xticks(())
plt.yticks(())

plt.show()
```
