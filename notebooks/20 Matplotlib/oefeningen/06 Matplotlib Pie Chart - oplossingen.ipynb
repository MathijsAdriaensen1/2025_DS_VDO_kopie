### Matplotlib Pie Chart - oplossingen

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import numpy as np
import pandas as pd
```

1. Write a  Python programming to display a pie chart of the popularity of programming Languages.
Sample data:

Programming languages:  
Java, Python, PHP, JavaScript, C#, C++  
Popularity: 22.2, 17.6, 8.8, 8, 7.7, 6.7  

The code snippet gives the output shown in the following screenshot:

```python
img = mpimg.imread('../../IMAGES/matplotlib-piechart-exercise-1.png')
plt.imshow(img)
plt.axis('off')  # Verberg de assen
plt.show()
```

```python
# Solution: 

# Data to plot
languages = ['Java', 'Python', 'PHP', 'JavaScript', 'C#', 'C++']
popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]
colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#9467bd", "#8c564b"]
explode = (0.1, 0, 0, 0, 0, 0)  # Explode the 1st slice

# Create figure and axis objects
fig, ax = plt.subplots()

# Create pie chart
ax.pie(popularity, explode=explode, labels=languages, colors=colors,
       autopct='%1.1f%%', shadow=True, startangle=140)

# Equal aspect ratio ensures that pie is drawn as a circle.
ax.axis('equal')  

# Show the plot
plt.show()
```

2. Write a Python programming to create a pie chart with a title of the popularity of programming Languages.
Sample data:

Programming languages:  
Java, Python, PHP, JavaScript, C#, C++  
Popularity: 22.2, 17.6, 8.8, 8, 7.7, 6.7  

The code snippet gives the output shown in the following screenshot:

```python
img = mpimg.imread('../../IMAGES/matplotlib-piechart-exercise-2.png')
plt.imshow(img)
plt.axis('off')  # Verberg de assen
plt.show()
```

```python
# Solution:

# Data to plot
languages = ['Java', 'Python', 'PHP', 'JavaScript', 'C#', 'C++']
popularity = [22.2, 17.6, 8.8, 8, 7.7, 6.7]
colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#9467bd", "#8c564b"]
explode = (0.1, 0, 0, 0, 0, 0)  # Explode the 1st slice

# Create figure and axis objects
fig, ax = plt.subplots()

# Create pie chart
ax.pie(popularity, explode=explode, labels=languages, colors=colors,
       autopct='%1.1f%%', shadow=True, startangle=140)

# Set title with a bounding box
ax.set_title("Popularity of Programming Languages\nWorldwide, Oct 2017 compared to a year ago", 
             bbox={'facecolor': '0.8', 'pad': 5})

# Equal aspect ratio ensures that pie is drawn as a circle.
ax.axis('equal')

# Show the plot
plt.show()
```

4. Write a Python programme to create a pie chart of gold medal achievements of five most successful countries in 2016 Summer Olympics. Read the data from a csv file.
   
Sample data:  
medal.csv  
country,gold_medal  
United States,46  
Great Britain,27  
China,26  
Russia,19  
Germany,17  

The code snippet gives the output shown in the following screenshot:

```python
# Solution:

# Wegschrijven van de data
with open('medal.csv','w') as my_file:
    my_file.write("""country,gold_medal
    United States,46
    Great Britain,27
    China,26
    Russia,19
    Germany,17
    """)

# Data inladen
df = pd.read_csv('medal.csv', sep=',', header=0)

# Data selecteren
country_data = df["country"]
medal_data = df["gold_medal"]
colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#8c564b"]
explode = (0.1, 0, 0, 0, 0)  # De eerste slice 'explode'

# Creëer figuur en assen objecten
fig, ax = plt.subplots()

# Maak taartdiagram
ax.pie(medal_data, labels=country_data, explode=explode, colors=colors,
       autopct='%1.1f%%', shadow=True, startangle=140)

# Titel instellen
ax.set_title("Gold medal achievements of five most successful\n" + 
             "countries in 2016 Summer Olympics")

# Zorg dat de taartdiagram als een cirkel getekend wordt
ax.axis('equal')

# Toon de plot
plt.show()
```

```python
# Solution

# Olympic color palette
colors = ["#0081C8", "#FCB131", "#000000", "#00A651", "#EE334E"]

# Exploding the United States slice for emphasis
explode = (0.1, 0, 0, 0, 0)  

# Creëer figuur en assen objecten
fig, ax = plt.subplots()

# Maak taartdiagram
ax.pie(medal_data, labels=country_data, explode=explode, colors=colors,
       autopct='%1.1f%%', shadow=False, startangle=140,
       wedgeprops={'edgecolor': 'white', 'linewidth': 1.5})

# Titel instellen
ax.set_title("Gold medal achievements of five most successful\n" + 
             "countries in 2016 Summer Olympics")

# Zorg dat de taartdiagram als een cirkel getekend wordt
ax.axis('equal')

# Toon de plot
plt.show()

# Save! 
fig.savefig('../../IMAGES/matplotlib-barchart-exercise-5.png')
```
