#### Maken van een plot

Om een plot te creëren moeten we een figuur en een reeks assen aanmaken.
Elke set assen is eigenlijk een plot, en een figure kan één of meerdere sets assen bevatten. 

De eenvoudigste manier om een figuur en assen te creëren is door de `subplots()` functie in de `pyplot` module te gebruiken.  Het figure object is een container die alles wat je ziet op de pagina omvat.  Het axes is het deel van de pagina wat de data bevat.  Het is het deel waarop we tekenen om onze data te visualiseren.

```python
fig, ax = plt.subplots()
```

Dus dit maakt een figuur en een set assen, maar natuurlijk is het leeg.

Om iets te laten verschijnen, moeten een functie of datapunten specificeren om op de assen te plotten.

```python
x_pts = np.linspace(-2 * np.pi, 2 * np.pi, 300)
y_pts = np.sin(x_pts) # gevectoriseerde numpy functie
ax.plot(x_pts, y_pts, label='sin')  # label voor de plot is 'sin'

# toon de legende
ax.legend()
```

```python
fig
```

Meestal maken en specificeren we de plots in één cel:

```python
fig, ax = plt.subplots()
ax.plot(x_pts, y_pts, label='sin')
ax.legend()
```

We kunnen meer grafieken toevoegen aan dezelfde assen, door eenvoudigweg meer grafieken te maken 'op' hetzelfde ax object:

```python
ax.plot(x_pts, np.cos(x_pts), label='cos')
ax.legend()
fig
```

We kunnen labels toevoegen aan de assen:

```python
ax.set_xlabel('radialen')
ax.set_ylabel('goniometrische functie')
ax.legend()
fig
```

```python
ax.set_title('Trig Functions')
fig
```
