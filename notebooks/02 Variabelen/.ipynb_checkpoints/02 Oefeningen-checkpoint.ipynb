#### Oefening 1: 
Schrijf een Python-programma dat een geheel getal (n) accepteert en de waarde van n + nn + nnn berekent. Voorbeeldwaarde van n is 5.
Verwachte uitkomst: 615.

```python
# Oplossing
# Vraag de gebruiker om een geheel getal in te voeren
n = int(input("Voer een geheel getal in: "))

# Bereken de waarde van n + nn + nnn
result = n + int(str(n) * 2) + int(str(n) * 3)
print(result)
```
