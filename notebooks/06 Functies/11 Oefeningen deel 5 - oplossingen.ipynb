### Oplossingen

```python
def isLeapYear(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)
```

```python
def numberOfDaysIn(year, month = 0):
    numberOfDaysIn = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if isLeapYear(year): numberOfDaysIn[2] = 29
    if month != 0: return numberOfDaysIn[month]
    else: return 337 + numberOfDaysIn[2]
```

```python
def numberOfDay(year, month, day, sinceYear = -1):
    numberOfDays = day
    for m in range(1, month):
        numberOfDays += numberOfDaysIn(year, m)
    if sinceYear != -1:
        for y in range(sinceYear, year):
            numberOfDays += numberOfDaysIn(y)
    return numberOfDays
```

```python
def date(days, sinceYear):
    if days >= 0:
        year = sinceYear
        month = 1
        diff = 1
    else:
        year = sinceYear - 1
        month = 12
        diff = -1
        days = -days
    while days > numberOfDaysIn(year):
        days -= numberOfDaysIn(year)
        year += diff
    while days > numberOfDaysIn(year, month):
        days -= numberOfDaysIn(year, month)
        month += diff
    if diff == -1:
        days = numberOfDaysIn(year, month) - days + 1
    return year, month, days
```

```python
def add(year, month, day, days):
    serialDays = numberOfDay(year, month, day, year) + days
    if serialDays <= 0:
        serialDays -= 1
    return date(serialDays, year)
```

```python
def before(aYear, aMonth, aDay, anotherYear, anotherMonth, anotherDay):
    res = dateDiffInDays(aYear, aMonth, aDay, anotherYear, anotherMonth, anotherDay)
    if res != 0: res = res / abs(res)
    return res
```

```python
def after(aYear, aMonth, aDay, anotherYear, anotherMonth, anotherDay):
    return -before(aYear, aMonth, aDay, anotherYear, anotherMonth, anotherDay)
```

```python
def dateDiffInDays(fromYear, fromMonth,  fromDay, toYear, toMonth, toDay):
    since = min(fromYear, toYear)
    return numberOfDay(toYear, toMonth, toDay, since) - numberOfDay(fromYear, fromMonth, fromDay, since)
```

```python
def birthdayAlreadyPassedThisYear(bornInYear, bornInMonth, bornOnDay):
    thisYear, thisMonth,thisDay = today()
    if thisYear < bornInYear: return False
    if before(thisYear, bornInMonth, bornOnDay, thisYear, thisMonth, thisDay) > 0:
        return True
    else:
        return False
```

```python
def age(bornInYear, bornInMonth, bornOnDay):
    thisYear, thisMonth, thisDay = today()
    currAge = thisYear - bornInYear
    if not birthdayAlreadyPassedThisYear(bornInYear, bornInMonth, bornOnDay):
        currAge -= 1
    if currAge < 0: currAge = 0
    return currAge
```
