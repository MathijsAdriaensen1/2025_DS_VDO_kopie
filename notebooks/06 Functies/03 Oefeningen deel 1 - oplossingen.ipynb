- Vraag de gebruiker om een ongekend aantal getallen, hou deze bij in een lijst en druk dan het volgende af: de som van alle getallen, welke getallen even zijn, welke getallen oneven zijn
- Maak een programma dat de gebruiker 5 namen en een salaris vraagt, is er geen salaris dan nemen we als default 2000 euro. Druk deze lijst dan af
- Schrijf een functie die bij elke iteratie één toevoegt aan het vorige resultaat (1, 2, 3, ...).  Test dit met 1000; print de output om de 100 iteraties.
- Schrijf een functie die een variabel aantal argumenten (met een minimum van één) aanneemt, en het gemiddelde van deze getallen teruggeeft.

- Schrijf een functie die twee strings krijgt als parameters. De functie retourneert het aantal tekens dat de strings gemeen hebben. Ieder teken wordt slechts eenmalig geteld, bijvoorbeeld, de strings "een" en "peer" hebben slechts één teken gemeen (de letter “e”). Je mag hoofdletters zien als verschillend van kleine letters. Merk op: de functie retourneert het aantal tekens dat de strings gemeen hebben, en moet het niet in de functie printen. Om de functie te testen kun je wel de retourwaarde printen in het hoofdprogramma.
- De Grerory-Leibnitz reeks benadert de waarde van π door de berekening van 4 ∗ ( 1/1 − 1/3 + 1/5 − 1/7 + 1/9... ) . Schrijf een functie die π benadert via deze reeks. De functie krijgt één parameter, namelijk een integer die aangeeft hoeveel van de termen tussen de haakjes in de reeks berekend moeten worden.

```python
def getallen_verwerken():
    getallen = []

    print("Voer getallen in (typ 'stop' om te stoppen):")
    while True:
        invoer = input("Getal: ")
        if invoer.lower() == 'stop':
            break
        try:
            getal = int(invoer)
            getallen.append(getal)
        except ValueError:
            print("Voer een geldig getal in of typ 'stop' om te stoppen.")

    som = sum(getallen)
    even_getallen = [g for g in getallen if g % 2 == 0]
    oneven_getallen = [g for g in getallen if g % 2 != 0]

    print(f"De som van alle getallen is: {som}")
    print(f"Even getallen: {even_getallen}")
    print(f"Oneven getallen: {oneven_getallen}")

# Roep de functie aan om het programma te starten
getallen_verwerken()
```

```python
def namen_en_salarissen():
    werknemers = []

    for i in range(5):
        naam = input(f"Voer de naam van werknemer {i + 1} in: ")
        salaris = input(f"Voer het salaris van {naam} in (of laat leeg voor 2000 euro): ")

        if salaris == '':
            salaris = 2000
        else:
            try:
                salaris = float(salaris)
            except ValueError:
                print("Ongeldig salaris, gebruik 2000 euro als standaard.")
                salaris = 2000

        werknemers.append((naam, salaris))

    print("\nWerknemerslijst:")
    for werknemer in werknemers:
        print(f"Naam: {werknemer[0]}, Salaris: {werknemer[1]} euro")

# Roep de functie aan om het programma te starten
namen_en_salarissen()
```

```python
def iteratieve_teller(tot_aantal):
    resultaat = 0
    for i in range(1, tot_aantal + 1):
        resultaat += 1
        if i % 100 == 0:
            print(f"Iteratie {i}: resultaat = {resultaat}")

# Test de functie met 1000 iteraties
iteratieve_teller(1000)
```

```python
def gemiddeld(*getallen):
    if len(getallen) == 0:
        raise ValueError("Minimaal één argument vereist")
    return sum(getallen) / len(getallen)

# Voorbeeldtest
print(gemiddeld(1, 2, 3, 4, 5))  # Output: 3.0
print(gemiddeld(10, 20, 30))     # Output: 20.0
```

```python
def tafel_van(getal):
    for i in range(1, 11):
        print(f"{i} * {getal} = {i * getal}")

# Voorbeeldtest
tafel_van(12)
```

```python
def gemeenschappelijke_tekens(string1, string2):
    set1 = set(string1)
    set2 = set(string2)
    gemeenschappelijk = set1 & set2
    return len(gemeenschappelijk)

# Voorbeeldtest
resultaat = gemeenschappelijke_tekens("een", "peer")  # Output: 1
print(resultaat)
```

```python
def benader_pi(aantal_termen):
    pi_benadering = 0
    teken = 1
    for i in range(aantal_termen):
        pi_benadering += teken * (4 / (2 * i + 1))
        teken *= -1
    return pi_benadering

# Voorbeeldtest
resultaat = benader_pi(1000)  # Benadering van π met 1000 termen
print(resultaat)
```
