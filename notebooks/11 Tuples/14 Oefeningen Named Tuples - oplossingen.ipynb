### Oefening 1

Schrijf een Python-programma dat een NamedTuple definieert genaamd Employee met velden zoals name, age en country. Print de name en country van elke werknemer.

```python
from collections import namedtuple
# Definieer de NamedTuple Employee
Employee = namedtuple("Employee", ["name", "age", "country"])

def main():
    try:
        # Maak een lijst van Employee-instanties
        employees = [
            Employee("Claerbout Olivier", 45, "BE"),
            Employee("Titus Andronicus", 30, "UK"),
            Employee("Gollem Merak", 28, "FR"),
            Employee("Nina Simone", 60, "USA"),
        ]
        
        # Print de naam en het land van elke werknemer
        for employee in employees:
            print("Employee Name:", employee.name)
            print("Employee Country:", employee.country)
            print()  # Print een lege regel voor scheiding
    except Exception as e:
        print("An error occurred:", e)

main()
```

Schrijf een Python-programma dat een dictionary maakt waarin de sleutels voedselnamen zijn en de waarden instanties van de Food-NamedTuple die de prijzen bevatten.

```python
# Definieer een NamedTuple genaamd 'Food' met de velden 'name' en 'price'
Food = namedtuple("Food", ["name", "price"])
# Maak instanties van de Food-NamedTuple
food1 = Food("Frieten met snack", 11.90)
food2 = Food("Pizza", 7.45)
food3 = Food("Slaatje", 5.45)
# Maak een woordenboek waarin de sleutels voedselnamen zijn en de waarden Food-instanties
food_dict = {
    food1.name: food1,
    food2.name: food2,
    food3.name: food3
}
# Toegang verkrijgen tot en printen van waarden met behulp van woordenboeksleutels
for food_name, food_info in food_dict.items():
    print(f"Food: {food_name}, Price: €{food_info.price:.2f}")
```

### Oefening 5

Schrijf een Python-functie die een namedtuple als invoer neemt met de name en price van het item. Laat de functie de name en price van het item teruggeven.

```python
# Definieer een NamedTuple genaamd 'Item' met de velden 'name' en 'price'
Item = namedtuple("Item", ["name", "price"])

def get_item_info(item):
    return item.name, item.price

# Maak instanties van de Item-NamedTuple
items = [
    Item(name="Mobile", price=300.00),
    Item(name="Desktop", price=1500.00),
    Item(name="Laptop", price=1200.00)
]

# Print de informatie van elk item
for item in items:
    item_name, item_price = get_item_info(item)
    print("Item Name:", item_name)
    print("Item Price:", item_price)
    print()
```

### Oefening 6

Schrijf een Python-functie die een Student-NamedTuple als argument neemt en het gemiddelde cijfer berekent.

```python
# Definieer een NamedTuple genaamd 'Student' met de velden 'name', 'age' en 'marks'
Student = namedtuple("Student", ["name", "age", "marks"])

def calculate_average_grade(student):
    # Bereken het totaal van de cijfers
    total_marks = sum(student.marks)
    # Bereken het gemiddelde cijfer
    average_grade = total_marks / len(student.marks)
    return average_grade

# Maak een instantie van de Student-NamedTuple
student1 = Student(name="Ain Ruth", age=22, marks=[89, 92, 75, 90, 86])

# Bereken en print het gemiddelde cijfer voor de student
average_grade = calculate_average_grade(student1)
print("Average Grade:", average_grade)
```

### Oefening 8

Schrijf een Python-programma dat een NamedTuple definieert genaamd Triangle met de velden side1, side2 en side3. Schrijf vervolgens een functie die een Triangle-NamedTuple als invoer neemt en de oppervlakte berekent.

```python
import math

# Definieer een Triangle-NamedTuple
Triangle = namedtuple('Triangle', ['side1', 'side2', 'side3'])

def valid_triangle(triangle):
    if (triangle.side1 + triangle.side2 > triangle.side3) and \
       (triangle.side2 + triangle.side3 > triangle.side1) and \
       (triangle.side1 + triangle.side3 > triangle.side2):
        return True
    else:
        return False

# Functie om de oppervlakte van een driehoek te berekenen
def triangle_area(triangle):
    if valid_triangle(triangle):
        print("Sides of the triangle:", triangle.side1, ',', triangle.side2, ',', triangle.side3)
        # Bereken de halve omtrek (de helft van de som van alle
        # zijden) van de driehoek
        s = (triangle.side1 + triangle.side2 + triangle.side3) / 2
        
        # Bereken de oppervlakte met behulp van de formule van Heron
        area = math.sqrt(s * (s - triangle.side1) * (s - triangle.side2) * (s - triangle.side3))
        return area
    else:
        return "Invalid triangle"

# Maak een instantie van de Triangle-NamedTuple
triangle_instance = Triangle(side1=1, side2=2, side3=7)

# Bereken en print de oppervlakte van de driehoek
area = triangle_area(triangle_instance)
print("Triangle Area:", area)
```

Schrijf een Python-programma dat een NamedTuple definieert genaamd Car met de velden make, model, year en engine (een NamedTuple die de details van de motor vertegenwoordigt). Maak een instantie van de Car-NamedTuple en print de attributen.

```python
# Definieer een NamedTuple voor Engine-details
Engine = namedtuple('Engine', ['type', 'cylinders'])

# Definieer een Car-NamedTuple
Car = namedtuple('Car', ['make', 'model', 'year', 'engine'])

# Maak een instantie van de Engine-NamedTuple
engine_instance = Engine(type='1.5L', cylinders=4)

# Maak een instantie van de Car-NamedTuple
car_instance = Car(make='Honda', model='City', year=2020, engine=engine_instance)

# Print de attributen van de Car-NamedTuple
print("Car Make:", car_instance.make)
print("Car Model:", car_instance.model)
print("Car Year:", car_instance.year)
print("Car Engine Type:", car_instance.engine.type)
print("Car Engine Cylinders:", car_instance.engine.cylinders)
```
