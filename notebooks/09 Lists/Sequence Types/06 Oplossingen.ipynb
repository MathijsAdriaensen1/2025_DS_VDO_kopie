#### Oefening 1

Gegeven de volgende string:

````python
s = 'FfEeDdCcBbAa'
````

Maak twee nieuwe variabelen aan die respectievelijk alleen de kleine letters en de hoofdletters van `s` bevatten, in de juiste alfabetische volgorde, d.w.z.:
- `'ABCDEF'`
- `'abcdef'`

##### Oplossing

De string `s` staat in omgekeerde volgorde, dus op een gegeven moment zullen we de string moeten omkeren. We zouden het kunnen doen voordat we de hoofd- en kleine letters extraheren, of we zouden het achteraf kunnen doen. Laten we het op beide manieren doen.

We beginnen eerst met het omkeren van de volgorde:

````python
reversed_s = s[::-1]
reversed_s
````

Nu kunnen we extended slicing gebruiken, beginnend bij `0`, om elke tweede karakter te selecteren:

````python
reversed_s[::2]
````

En beginnend bij `1` om de hoofdletters te selecteren:

````python
reversed_s[1::2]
````

Als alternatief zouden we eerst de hoofd-/kleine letters kunnen kiezen, en vervolgens het resultaat omkeren:

````python
s[::2][::-1]
````

````python
s[1::2][::-1]
````

#### Oefening 2

Voeg de volgende tuples samen tot één tuple, maar vervang de oneven waarden door nullen (`0`).

````python
t1 = 1, 2, 3, 4, 5, 6
t2 = 7, 8, 9, 10
t3 = 11, 12, 13, 14, 15, 16, 17
````

Je kunt ervan uitgaan dat elke tuple een reeks opeenvolgende integergetallen is die begint met een oneven integer.
Schrijf je code zo generiek als mogelijk.

##### Oplossing

We kunnen tuples niet muteren, dus zullen we eerst onze tuples naar lijsten moeten converteren:

````python
l1 = list(t1)
l2 = list(t2)
l3 = list(t3)
````

Nu we mutable sequenties hebben, kunnen we uitgebreide slicing gebruiken om de oneven getallen te vervangen door `0` in elke lijst:

````python
l1[::2] = [0, 0, 0]
l2[::2] = [0, 0]
l3[::2] = [0, 0, 0, 0]
````

````python
l1
````

````python
l2
````

````python
l3
````

Dus dit werkt, maar je zult merken dat we (in ons hoofd) moesten berekenen hoeveel nullen we moesten vervangen met de uitgebreide slice (aangezien bij uitgebreid slicen het aantal elementen aan beide kanten van de toewijzing overeen moet komen).
Dit is geen generieke code - in plaats hiervan kunnen we vaststellen hoeveel elementen er in elke extended slice zitten met behulp van de`len` functie:

````python
len(l1[::2]), len(l2[::2]), len(l3[::2])
````

We weten ook dat we eenvoudig een lijst van `n` herhaalde elementen kunnen maken door simpelweg een lijst met een geheel getal te vermenigvuldigen:

````python
[0] * 5
````

Dus, we zouden elke uitgebreide slicing kunnen vervangen door een lijst `[0]` vermenigvuldigd met de lengte van de uitgebreide slicing:

````python
l1 = list(t1)
l2 = list(t2)
l3 = list(t3)

l1[::2] = [0] * len(l1[::2])
l2[::2] = [0] * len(l2[::2])
l3[::2] = [0] * len(l3[::2])

l1, l2, l3
````

Nu kunnen we die lijsten samenvoegen:

````python
result = l1 + l2 + l3
result
````

Maar we willen eigenlijk een `tuple` voor ons resultaat:

````python
result = tuple(l1 + l2 + l3)
result
````

Alles samenvoegen:

````python
l1 = list(t1)
l2 = list(t2)
l3 = list(t3)

l1[::2] = [0] * len(l1[::2])
l2[::2] = [0] * len(l2[::2])
l3[::2] = [0] * len(l3[::2])

result = tuple(l1 + l2 + l3)
result
````

````python
# Of korter:
large_tuple = tuple([x if x%2 == 0 else 0 for x in (*t1,*t2,*t3)])
large_tuple
````

````python
lijst = []

for x in t1 + t2 + t3:
    x = x if x%2 == 0 else 0
    lijst.append(x)
    
tuple(lijst)        
````

#### Oefening 3  

Gegeven de volgende matrix:

````python
m = [
    [0, 0, 0],
    [0, 0, 0],
    [0, 0, 0]
]
````

Maak deze matrix tot een identiteitsmatrix (de diagonale elementen instellen op `1`).
Your code should *mutate* `m`.

##### Oplossing

Aangezien we `m` moeten muteren, kunnen we de waarden eenvoudig rechtstreeks toewijzen aan `m`:

````python
m[0][0] = 1
m[1][1] = 1
m[2][2] = 1
m
````

#### Oefening 4

Doe hetzelfde probleem als Oefening 3, maar muteer `m` **niet**.

##### Oplossing

We kunnen de bewerkingen hierboven niet rechtstreeks op `m` uitvoeren.  We proberen via een shallow copy:

````python
m = [[0, 0, 0], [0, 0, 0], [0, 0, 0]]
````

````python
result = m.copy()
result
````

````python
result[0][0] = 1
result[1][1] = 1
result[2][2] = 1

result
````

Dus `result` ziet er goed uit, maar hoe zit het met `m`?

````python
m
````

Deze is ook gewijzigd, aangezien de shallow kopie nog steeds verwijzingen naar de sublijsten in `m` bevatte.

Om hier omheen te werken, moeten we een **diepe** kopie maken.

````python
from copy import deepcopy
````

Laten we `m` terugbrengen naar zijn oorspronkele staat:

````python
m = [
    [0, 0, 0],
    [0, 0, 0],
    [0, 0, 0]
]
````

En laten we de diepe kopie maken:

````python
result = deepcopy(m)

result[0][0] = 1
result[1][1] = 1
result[2][2] = 1

result
````

En `m` blijft nu volledig onaangeroerd:

````python
m
````

#### Oefening 5

Je krijgt een lijst van tuples die elk 4 waarden bevatten:
```
(amount, currency, target_currency, exchange_rate)
```

````python
data = [
    (100, 'USD', 'EUR', 0.83),
    (100, 'USD', 'CAD', 1.27),
    (100, 'CAD', 'EUR', 0.65)
]
````

Schrijf code die de `amount` van zijn `currency` naar zijn `target_currency` converteert met behulp van de `exchange_rate` (die de wisselkoers is voor `1` `currency` in `target_currency`).
Maak je code zo algemeen mogelijk. 

Je zal hier drie blokken code nodig hebben die bijna identiek zijn. 

Gebruik unpacking om de waarden in elke tuple aan variabelen toe te wijzen.  Je resultaat per rij zou iets moeten geven als 
```
100 USD = 83 EUR
```

##### Oplossing

We zullen eerst met rij `0` omgaan, en daarna onze code herhalen voor de andere twee rijen.

````python
row = 0

amount, currency, target_currency, exchange_rate = data[row]
converted = amount * exchange_rate
print(amount, currency, '=', converted, target_currency, sep=' ')
````

Nu kunnen we dezelfde code gebruiken, door alleen de waarde van `row` te veranderen om de rest van onze resultaten te krijgen:

````python
row = 1

amount, currency, target_currency, exchange_rate = data[row]
converted = amount * exchange_rate
print(amount, currency, '=', converted, target_currency, sep=' ')
````

````python
row = 2

amount, currency, target_currency, exchange_rate = data[row]
converted = amount * exchange_rate
print(amount, currency, '=', converted, target_currency, sep=' ')
````
