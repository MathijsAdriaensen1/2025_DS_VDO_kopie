### Oefening 1: Dieren Simuleren
   - Maak vier dieren aan: een leeuw, giraf, kat, en hond.
   - Elk dier moet een geluid kunnen maken, een lengte hebben (klein, gemiddeld, groot, zeer groot) en een loopsnelheid (uitgedrukt in seconde per meter).
   - Plaats al deze dieren in een lijst en druk voor elk dier het type, de lengte, en de loopsnelheid af.

```python
class Dier:
    def __init__(self, naam, lengte, snelheid):
        self.naam = naam
        self.lengte = lengte
        self.snelheid = snelheid

    def maak_geluid():
        pass

    def details(self):
        return f"ik  ben {self.lengte} lang en loop {self.snelheid} snel."

    @property
    def lengte(self):
        return self._lengte
    
    @lengte.setter
    def lengte(self, value):
        if value in ['klein', 'gemiddeld', 'groot', 'zeer groot']:
            self._lengte = value
        else:
            print('Foutieve lengte ingegeven.  Kies uit klein, gemiddeld, groot, zeer groot.')


class Leeuw(Dier):
    def maak_geluid(self):
        return 'Roar'

class Giraf(Dier):
    def maak_geluid(self):
        return 'Whhhaaatt??'

class Kat(Dier):
    def maak_geluid(self):
        return 'Miauw'

class Hond(Dier):
    def maak_geluid(self):
        return 'Waf'

f = Hond('Fatale', 'gemiddeld', 30)
g = Giraf('Herman', 'zeer groot', 60)
m = Kat('Marcel', 'klein', 28)
j = Leeuw('Jules', 'groot', 70)
```

```python
dieren = [f, g, m, j]
for dier in dieren:
    print(dier.maak_geluid())
```
