# Oefening 1: Interface Ontwerpen

1. **Ontwerp een interface** genaamd `Shape` met de methoden `draw()` en `calculate_area()`.
2. **Implementeer de interface** in twee klassen: `Circle` en `Square`.

````python
## Oplossing:

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def draw(self):
        pass

    @abstractmethod
    def calculate_area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def draw(self):
        print("Drawing a Circle")

    def calculate_area(self):
        return 3.14 * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def draw(self):
        print("Drawing a Square")

    def calculate_area(self):
        return self.side ** 2

# Test de implementatie
circle = Circle(5)
square = Square(4)

circle.draw()
print("Circle Area:", circle.calculate_area())

square.draw()
print("Square Area:", square.calculate_area())
````

## **Probleemstelling**

Je werkt aan een data science-project waarin gegevens van tweedehandswagens verwerkt moeten worden. Deze gegevens komen in verschillende formaten binnen: **Excel**, **CSV**, en **JSON**. Je hebt als taak om een gestandaardiseerde data pipeline te bouwen die de gegevens uit elk formaat kan:

1. **Valideren** volgens specifieke regels.
2. **Transformeren** naar een uniform CSV-formaat.
3. **Rapporteren** over de resultaten van de verwerking.

Om dit te realiseren gebruik je een **interface** om de structuur te definiëren van de datastructuurverwerking. Elke implementatie volgt dezelfde interface, maar bevat specifieke logica voor het verwerken van een bepaald bestandsformaat.

---

## **Specificaties**

### **Interface**

Je bouwt een abstracte klasse die de volgende methoden definieert:

- **`load_data(file_path: str)`**:

  - Laadt de data vanuit een specifiek bestandsformaat.

- **`validate_data()`**:

  - Controleert de data op volledigheid en validiteit.

- **`transform_data(output_path: str)`**:

  - Transformeert de data naar een uniform CSV-formaat en slaat het bestand op.

### **Validatieregels**

1. **Merk**:

   - Het merk van de wagen moet gekend zijn (car_brands_and_models.json).
   - Fuzzy matching is een techniek die kleine fouten in data herkent en corrigeert door overeenkomsten te berekenen tussen strings, zoals met de Levenshtein-afstand. Gebruik deze techniek om schrijffouten in het merk te corrigeren en te mappen op het juiste merk met **fuzzy matching**.

2. **Wagen Type**:

   - Het type moet overeenkomen met een bestaand type voor dit merk.
   - Gebruik ook hier fuzzy matching om het type te corrigeren bij schrijffouten.

3. **Cilinderinhoud**:

   - De waarde moet liggen tussen **1000** en **3600 cc**.

4. **Bouwjaar**:

   - Het bouwjaar moet liggen tussen **1990** en het huidige jaar.

5. **Brandstof**:

   - Het brandstoftype moet worden omgerekend naar **dummy-variabelen** (‘benzine’, ‘diesel’, ‘lpg’, ‘elektrisch’).
   - Wagens met andere brandstoftypes komen zo weinig voor dat we deze niet in het model hebben opgenomen.

6. **Kilometerstand**:

   - De kilometerstand moet liggen tussen **0** en **300.000**.

### **Transformaties**

1. **String Normalisatie**:

   - Alle stringvelden moeten worden genormaliseerd naar lowercase en speciale tekens moeten worden verwijderd voor je fuzzy matching toepast.

2. **Brandstof Dummy Encoding**:

   - Maak handmatig dummy-kolommen voor de verschillende brandstoftypes. Dummy encoding is een techniek waarbij categorische gegevens worden omgezet in meerdere binaire kolommen (dummies). Voor elke categorie wordt een kolom gemaakt met waarden 0 of 1, afhankelijk van of de categorie van toepassing is.

3. **ID’s voor Merken**:

   - Zet het merk om naar een unieke ID in plaats van de naam in tekst.

4. **Rapportage**:

   - Genereer een rapport (TXT-bestand) met:
     - Aantal rijen in het originele bestand.
     - Aantal correct verwerkte rijen.
     - Aantal fouten per type controle.

````python
class DataProcessor(ABC):
    """
    Abstract class for data processing. Defines methods that must be implemented
    by all subclasses to process various file types.
    """
    @abstractmethod
    def load_data(self, file_path: str):
        """Load data from a specific file."""
        pass

    @abstractmethod
    def validate_data(self):
        """Validate data according to specific rules."""
        pass

    @abstractmethod
    def transform_data(self, output_path: str):
        """Transform data to a uniform format and save to CSV."""
        pass
````

````python
class ExcelProcessor(DataProcessor):
    def __init__(self, brand_model_file: str):
        self.data = None
        self.report = {}
        self.brand_model_data = self.load_brand_model_data(brand_model_file)

    @staticmethod
    def load_brand_model_data(file_path: str):
        """Load brand and model data from a JSON file."""
        with open(file_path, 'r') as file:
            return json.load(file)

    def load_data(self, file_path: str):
        """Load data from an Excel file."""
        self.data = pd.read_excel(file_path)
        self.report['original_rows'] = len(self.data)

    def validate_data(self):
        """Validate data for completeness and accuracy."""
        self.report['validation_errors'] = {}
        current_year = datetime.datetime.now().year

        # Validate year
        self.data['valid_year'] = self.data['year'].apply(
            lambda x: 1990 <= x <= current_year if not pd.isnull(x) else False
        )
        # Inzake reporting: we willen enkel weten hoeveel records NIET ok zijn dus enkel de invalid years...
        # gebruik ~ : True wordt False en omgekeerd -> een logische NOT operator.
        self.report['validation_errors']['invalid_years'] = len(self.data[~self.data['valid_year']])

        # Match brands directly
        self.data['matched_brand'] = self.data['brand'].apply(lambda x: x if x in self.brand_model_data else None))
        
        # Fuzzy matching voor de brands die leeg zijn
        self.data.loc[self.data['matched_brand'].isnull(), 'matched_brand'] = self.data['brand'].apply(
            lambda x: process.extractOne(x.lower(), self.brand_model_data.keys())[0]  # -> [0]: retourneer de beste match
            # Vergelijkt x.lower() (de merknaam) met de sleutels van self.brand_model_data (een verzameling merkmodellen)
            if pd.notnull(x) else None
            # Controleert of de waarde x in de brand-kolom niet leeg (NaN) is; als x leeg is, retourneert de lambda None
        )
        self.data['brand_match_score'] = self.data['brand'].apply(
            lambda x: 100 if x in self.brand_model_data else
            process.extractOne(x.lower(), self.brand_model_data.keys())[1]
            if pd.notnull(x) else 0
        )
        self.report['validation_errors']['invalid_brands'] = len(self.data[self.data['brand_match_score'] < 80])

        # Validate model using matched brand
        def validate_model(row):
            brand = row['matched_brand']
            model = row['model']
            if pd.isnull(brand) or pd.isnull(model):
                return None, 0
            models = self.brand_model_data.get(brand, [])
            match = process.extractOne(model.lower(), models) if models else (None, 0)
            return match[0], match[1]

        self.data[['matched_model', 'model_match_score']] = self.data.apply(
            validate_model, axis=1, result_type='expand'
        )
        self.report['validation_errors']['invalid_models'] = len(self.data[self.data['model_match_score'] < 80])

        # Validate kilometerstand
        self.data['valid_kilometers'] = self.data['kilometers'].apply(
            lambda x: 0 <= x <= 300000 if not pd.isnull(x) else False
        )
        self.report['validation_errors']['invalid_kilometers'] = len(self.data[~self.data['valid_kilometers']])

    def transform_data(self, output_path: str):
        """Transform and save data to CSV."""
        # Normalize strings
        self.data['normalized_brand'] = self.data['matched_brand'].str.lower().str.strip()
        self.data['normalized_model'] = self.data['matched_model'].str.lower().str.strip()

        # Add dummy variables for fuel type
        fuel_dummies = pd.get_dummies(self.data['fuel_type'], prefix='fuel')
        self.data = pd.concat([self.data, fuel_dummies], axis=1)

        # Save to CSV
        self.data.to_csv(output_path, index=False)
        self.report['processed_rows'] = len(self.data)

    def generate_report(self, report_path: str):
        """Generate a processing report."""
        with open(report_path, 'w') as report_file:
            for key, value in self.report.items():
                report_file.write(f"{key}: {value}\n")
````

````python
class CSVProcessor(ExcelProcessor):
    def load_data(self, file_path: str):
        """Load data from a CSV file."""
        self.data = pd.read_csv(file_path)
        self.report['original_rows'] = len(self.data)


class JSONProcessor(ExcelProcessor):
    def load_data(self, file_path: str):
        """Load data from a JSON file."""
        self.data = pd.read_json(file_path)
        self.report['original_rows'] = len(self.data)

````

````python
def process_file(processor: DataProcessor, file_path: str, output_path: str, report_path: str):
    """
    Process a file using a specified data processor.
    Args:
        processor (DataProcessor): The processor object to use.
        file_path (str): Path to the input file.
        output_path (str): Path to save the transformed data.
        report_path (str): Path to save the processing report.
    """
    processor.load_data(file_path)
    processor.validate_data()
    processor.transform_data(output_path)
    processor.generate_report(report_path)


if __name__ == "__main__":
    os.makedirs("reports", exist_ok=True)
    os.makedirs("processed_data", exist_ok=True)

    # Load the brand and model mapping file
    brand_model_file = "DATA/car_brands_and_models.json"

    # Process Excel file
    excel_processor = ExcelProcessor(brand_model_file)
    process_file(
        excel_processor,
        "DATA/used_cars_dataset.xlsx",
        "DATA/processed/processed_cars_from_excel.csv",
        "DATA/reports/excel_processing_report.txt"
    )

    # Process CSV file
    csv_processor = CSVProcessor(brand_model_file)
    process_file(
        csv_processor,
        "DATA/used_cars_dataset.csv",
        "DATA/processed/processed_cars_from_csv.csv",
        "DATA/reports/csv_processing_report.txt"
    )

    # Process JSON file
    json_processor = JSONProcessor(brand_model_file)
    process_file(
        json_processor,
        "DATA/used_cars_dataset.json",
        "DATA/processed/processed_cars_from_json.csv",
        "DATA/reports/json_processing_report.txt"
    )
````
