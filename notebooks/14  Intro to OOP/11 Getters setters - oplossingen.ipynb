### **1. Bankrekening Klasse**
**Opgave**:  
Maak een klasse `Bankrekening` met de volgende specificaties:  
1. **Eigenschappen**:
   - `rekeningnummer`: Alleen-lezen eigenschap (getter) die het rekeningnummer teruggeeft.
   - `saldo`: Lees- en schrijfbare eigenschap met een getter en setter. De setter moet controleren dat het saldo niet negatief kan worden.

2. **Constructor**:
   - De constructor moet het rekeningnummer en een beginnend saldo (standaardwaarde 0) accepteren. Als het meegegeven saldo negatief is, moet het automatisch naar 0 worden gezet.

3. **Validatie**:
   - Bij een poging om een negatief saldo in te stellen, moet een foutmelding worden gegenereerd.

Test je implementatie door een object te maken, het saldo te wijzigen, en foutieve waarden uit te proberen.

```python
class Bankrekening:
    def __init__(self, rekeningnummer, saldo):
        self._rekeningnummer = rekeningnummer  # Privé-eigenschap
        self._saldo = saldo if saldo >= 0 else 0  # Directe validatie

    @property
    def rekeningnummer(self):
        return self._rekeningnummer  # Alleen lezen

    @property
    def saldo(self):
        return self._saldo

    @saldo.setter
    def saldo(self, waarde):
        if waarde >= 0:
            self._saldo = waarde
        else:
            raise ValueError("Saldo mag niet negatief zijn!")

# Test
rekening = Bankrekening("BE123456789", 100)
print(rekening.rekeningnummer)  # Output: BE123456789
print(rekening.saldo)  # Output: 100
rekening.saldo = 200
print(rekening.saldo)  # Output: 200
```

### **2. Student Klasse**
**Opgave**:  
Maak een klasse `Student` met de volgende specificaties:  
1. **Eigenschappen**:
   - `naam`: Lees- en schrijfbare eigenschap (getter en setter). De setter moet controleren dat de naam een niet-lege string is.
   - `studentnummer`: Lees- en schrijfbare eigenschap. De setter moet controleren dat het studentnummer alleen cijfers bevat.
   - `gemiddelde`: Lees- en schrijfbare eigenschap. De setter moet controleren dat de waarde tussen 0 en 10 ligt.

2. **Constructor**:
   - De constructor moet een naam, studentnummer, en een beginnend gemiddelde accepteren.

3. **Validatie**:
   - Een foutmelding moet worden gegenereerd als een ongeldige waarde wordt ingevoerd voor een van de eigenschappen.

Test je implementatie door een object te maken, eigenschappen te wijzigen, en foutieve waarden in te stellen.

```python
class Student:
    def __init__(self, naam, studentnummer, gemiddelde):
        self._naam = naam
        self._studentnummer = studentnummer
        self.gemiddelde = gemiddelde  # Roept setter aan

    @property
    def naam(self):
        return self._naam

    @naam.setter
    def naam(self, waarde):
        if isinstance(waarde, str) and waarde:
            self._naam = waarde
        else:
            raise ValueError("Ongeldige naam!")

    @property
    def studentnummer(self):
        return self._studentnummer

    @studentnummer.setter
    def studentnummer(self, waarde):
        if isinstance(waarde, str) and waarde.isdigit():
            self._studentnummer = waarde
        else:
            raise ValueError("Studentnummer moet een geldig nummer zijn!")

    @property
    def gemiddelde(self):
        return self._gemiddelde

    @gemiddelde.setter
    def gemiddelde(self, waarde):
        if 0 <= waarde <= 10:
            self._gemiddelde = waarde
        else:
            raise ValueError("Gemiddelde moet tussen 0 en 10 liggen!")

# Test
student = Student("Jan", "12345", 8.5)
print(student.gemiddelde)  # Output: 8.5
student.gemiddelde = 9.0
print(student.gemiddelde)  # Output: 9.0
```

### **3. Auto Klasse**
**Opgave**:  
Ontwerp een klasse `Auto` met de volgende specificaties:  
1. **Eigenschappen**:
   - `merk`: Lees- en schrijfbare eigenschap zonder extra validatie.
   - `model`: Lees- en schrijfbare eigenschap zonder extra validatie.
   - `kilometerstand`: Lees- en schrijfbare eigenschap. De setter moet controleren dat de kilometerstand niet lager wordt dan de huidige waarde.

2. **Constructor**:
   - De constructor moet een merk, model, en een optionele beginnende kilometerstand (standaard 0) accepteren.

3. **Validatie**:
   - Bij een poging om de kilometerstand te verlagen, moet een foutmelding worden gegooid.

Test je implementatie door een auto te maken, de kilometerstand te verhogen, en foutieve waarden in te stellen.

```python
class Auto:
    def __init__(self, merk, model, kilometerstand=0):
        self._merk = merk
        self._model = model
        self._kilometerstand = kilometerstand

    @property
    def merk(self):
        return self._merk

    @merk.setter
    def merk(self, waarde):
        self._merk = waarde

    @property
    def model(self):
        return self._model

    @model.setter
    def model(self, waarde):
        self._model = waarde

    @property
    def kilometerstand(self):
        return self._kilometerstand

    @kilometerstand.setter
    def kilometerstand(self, waarde):
        if waarde >= self._kilometerstand:
            self._kilometerstand = waarde
        else:
            raise ValueError("Kilometerstand kan niet worden verlaagd!")

# Test
auto = Auto("Toyota", "Corolla", 50000)
print(auto.kilometerstand)  # Output: 50000
auto.kilometerstand = 60000
print(auto.kilometerstand)  # Output: 60000
```

### **4. Personeelslid Klasse**
**Opgave**:  
Ontwerp een klasse `Personeelslid` met de volgende specificaties:  
1. **Eigenschappen**:
   - `naam`: Alleen-lezen eigenschap (getter).
   - `positie`: Lees- en schrijfbare eigenschap. De setter moet controleren dat de waarde een niet-lege string is.
   - `salaris`: Lees- en schrijfbare eigenschap. De setter moet controleren dat het salaris niet lager is dan een vastgesteld minimumloon (bijvoorbeeld €2000).

2. **Constructor**:
   - De constructor moet een naam, positie, en beginnend salaris accepteren.

3. **Validatie**:
   - Als een salaris wordt ingevoerd dat lager is dan het minimumloon, moet een foutmelding worden gegooid.

Test je implementatie door een personeelslid aan te maken, eigenschappen te wijzigen, en foutieve waarden in te stellen.

```python
class Personeelslid:
    MINIMUMLOON = 2000

    def __init__(self, naam, positie, salaris):
        self._naam = naam
        self._positie = positie
        self.salaris = salaris  # Roept setter aan

    @property
    def naam(self):
        return self._naam

    @property
    def positie(self):
        return self._positie

    @positie.setter
    def positie(self, waarde):
        if isinstance(waarde, str) and waarde:
            self._positie = waarde
        else:
            raise ValueError("Positie moet een geldige string zijn!")

    @property
    def salaris(self):
        return self._salaris

    @salaris.setter
    def salaris(self, waarde):
        if waarde >= self.MINIMUMLOON:
            self._salaris = waarde
        else:
            raise ValueError(f"Salaris moet minstens {self.MINIMUMLOON} zijn!")

# Test
personeel = Personeelslid("Anna", "Manager", 3000)
print(personeel.salaris)  # Output: 3000
personeel.salaris = 2500
print(personeel.salaris)  # Output: 2500
personeel.naam = 'Olivier'
```

### **5. TemperatuurSensor Klasse**
**Opgave**:  
Ontwerp een klasse `TemperatuurSensor` met de volgende specificaties:  
1. **Eigenschappen**:
   - `temperatuur_celsius`: Lees- en schrijfbare eigenschap die de temperatuur in graden Celsius opslaat.
   - `temperatuur_fahrenheit`: Lees- en schrijfbare eigenschap die de temperatuur in Fahrenheit teruggeeft en accepteert. Interne opslag blijft in Celsius.

2. **Constructor**:
   - De constructor moet een beginnende temperatuur in graden Celsius accepteren.

3. **Omrekening**:
   - Zorg ervoor dat de setter voor `temperatuur_fahrenheit` de waarde correct omzet naar Celsius voor interne opslag.

Test je implementatie door een sensor aan te maken, de temperatuur in Celsius en Fahrenheit te wijzigen, en de correcte omzetting te controleren.

```python
class TemperatuurSensor:
    def __init__(self, temperatuur_celsius):
        self.temperatuur_celsius = temperatuur_celsius  # Roept setter aan

    @property
    def temperatuur_celsius(self):
        return self._temperatuur

    @temperatuur_celsius.setter
    def temperatuur_celsius(self, waarde):
        self._temperatuur = waarde

    @property
    def temperatuur_fahrenheit(self):
        return self._temperatuur * 9/5 + 32

    @temperatuur_fahrenheit.setter
    def temperatuur_fahrenheit(self, waarde):
        self._temperatuur = (waarde - 32) * 5/9

# Test
sensor = TemperatuurSensor(25)
print(sensor.temperatuur_celsius)  # Output: 25
print(sensor.temperatuur_fahrenheit)  # Output: 77.0
sensor.temperatuur_fahrenheit = 86
print(sensor.temperatuur_celsius)  # Output: 30.0
```
