1. Opgave: Maak een klasse genaamd Boek die drie attributen bevat: titel, auteur, en aantal_paginas. Elk boekobject moet deze eigenschappen hebben. Maak 2 boeken aan en vraag de eigenschappen ervan op.

```python
class Boek:
    def __init__(self, titel, auteur, aantal_paginas):
        self.titel = titel
        self.auteur = auteur
        self.aantal_paginas = aantal_paginas

pythonboek = Boek('Python voor Data Science', 'Guido Van Rossum', 1245)
sqlboek = Boek('SQL voor data analyse', 'Tim Hellemans', 365)

print(f"Titel: {pythonboek.titel}")
print(f"Auteur: {pythonboek.auteur}")
print(f"Aantal pagina's: {pythonboek.aantal_paginas}")
```

2. Opgave: Maak een klasse genaamd Student die drie attributen bevat: naam, leeftijd, en studierichting. Maak 3 personen aan, en stop deze objecten in een Python list.  Loop over de list en print voor elke student de gegevens af.

```python
class Student:
    def __init__(self, naam, leeftijd, studierichting):
        self.naam = naam
        self.leeftijd = leeftijd
        self.studierichting = studierichting

leno = Student('Leno', 16, 'Lawi')
tito = Student('Tito', 11, 'alg')
rusti = Student('Rustine', 6, 'Dressuur')

studenten = [leno, tito, rusti]

for student in studenten:
    print(f"Naam: {student.naam}")
    print(f"Leeftijd: {student.leeftijd}")
    print(f"Studierichting: {student.studierichting}\n")
```

3. Vraag aan de gebruiker zijn voornaam, familienaam en leeftijd. Zet dit in een object, druk nadien deze waarden af

```python
class Persoon:
    def __init__(self, voornaam, familienaam, leeftijd):
        self.voornaam = voornaam
        self.familienaam = familienaam
        self.leeftijd = leeftijd

    def begroet(self):
        print(f"Hallo {self.voornaam}")

    def __repr__(self):
        return f"Ik ben {self.voornaam} en ik ben {self.leeftijd} jaar oud."

voornaam = input("Voer je voornaam in: ")
familienaam = input("Voer je familienaam in: ")
leeftijd = int(input("Voer je leeftijd in: "))

# Maak een Persoon object en druk de informatie af
persoon = Persoon(voornaam, familienaam, leeftijd)
print(f"Voornaam: {persoon.voornaam}")
print(f"Familienaam: {persoon.familienaam}")
print(f"Leeftijd: {persoon.leeftijd}")
```

4. Vraag aan de gebruiker minstens 2 wagens waar hij mij gereden heeft. Elke wagen heeft een kleur/merk/nummerplaat. Druk alle wagens af

```python
class Wagen:
    def __init__(self, kleur, merk, nummerplaat):
        self.kleur = kleur
        self.merk = merk
        self.nummerplaat = nummerplaat

    def toon_informatie(self):
        return f"{self.merk} ({self.kleur}) met nummerplaat: {self.nummerplaat}"

# Functie om wagens van de gebruiker op te vragen
def vraag_wagens_op():
    wagens = []

    while True:
        kleur = input("Voer de kleur van de wagen in: ")
        merk = input("Voer het merk van de wagen in: ")
        nummerplaat = input("Voer de nummerplaat van de wagen in: ")

        wagens.append(Wagen(kleur, merk, nummerplaat))

        nog_een_wagen = input("Wil je nog een wagen invoeren? (ja/nee): ").strip().lower()
        if nog_een_wagen != "ja":
            if len(wagens) < 2:
                print("Je moet minstens 2 wagens invoeren.")
            else:
                break

    return wagens

# Vraag de wagens van de gebruiker op
alle_wagens = vraag_wagens_op()

# Druk alle ingevoerde wagens af
print("\nJe hebt de volgende wagens ingevoerd:")
for wagen in alle_wagens:
    print(wagen.toon_informatie())
```

5. Vraag de gebruiker om een paar studenten in te geven. Een student kan kiezen uit een paar richtingen : programmeren, bouw, electriciteit. Druk dan alle studenten en hun richting af

```python
class Student:
    def __init__(self, naam, studierichting):
        self.naam = naam
        self.studierichting = studierichting

    def toon_informatie(self):
        return f"Student: {self.naam}, Studierichting: {self.studierichting}"

# Functie om studenten van de gebruiker op te vragen
def vraag_studenten_op():
    studenten = []
    richtingen = ["programmeren", "bouw", "electriciteit"]

    while True:
        naam = input("Voer de naam van de student in: ")

        # Laat de gebruiker een geldige studierichting kiezen
        while True:
            studierichting = input("Kies de studierichting (programmeren, bouw, electriciteit): ").strip().lower()
            if studierichting in richtingen:
                break
            else:
                print("Ongeldige keuze. Kies uit: programmeren, bouw, electriciteit.")

        studenten.append(Student(naam, studierichting))

        nog_een_student = input("Wil je nog een student invoeren? (ja/nee): ").strip().lower()
        if nog_een_student != "ja":
            break

    return studenten

# Vraag de studenten van de gebruiker op
alle_studenten = vraag_studenten_op()

# Druk alle ingevoerde studenten af
print("\nJe hebt de volgende studenten ingevoerd:")
for student in alle_studenten:
    print(student.toon_informatie())
```
