### **Oefening 1: LEGB-regel in een Methode**
Maak een class `Berekening` met een methode `som` die twee getallen optelt. Volg deze stappen:
1. Definieer een lokale variabele `x` binnen de methode.
2. Maak een andere variabele `x` op klasse-niveau.
3. Gebruik `print` om te laten zien welke `x` wordt gebruikt in de methode.

**Test:**  
- Maak een object van `Berekening` en voer de methode `som` uit. Begrijp hoe Python kiest welke `x` wordt gebruikt (local vs. global scope).

```python
# oplossing

class Berekening:
    # Klasse-niveau variabele
    x = 100

    def som(self, a, b):
        # Lokale variabele
        x = 10
        print(f"Lokale x binnen de methode: {x}")
        print(f"Klasse-niveau x: {Berekening.x}")
        return a + b

# Test de class
berekening = Berekening()

# Roep de methode som aan en bekijk het gedrag van x
resultaat = berekening.som(5, 7)
print(f"Resultaat van de som: {resultaat}")

# Toegang tot de klasse-niveau variabele
print(f"Klasse-niveau x buiten de methode: {Berekening.x}")
```

### **Oefening 2: Gebruik van `global`**
Maak een globale variabele `TELLER` en een class `TelMachine` met:
1. Een methode `verhoog` die `TELLER` verhoogt met 1. Gebruik `global` om de waarde aan te passen.
2. Een methode `toon_teller` om de huidige waarde van `TELLER` af te drukken.

**Test:**  
- Roep meerdere keren `verhoog` aan en controleer de waarde van `TELLER` met `toon_teller`.

```python
# oplossing

# Globale variabele
TELLER = 0

class TelMachine:
    def verhoog(self):
        global TELLER  # Gebruik de globale variabele
        TELLER += 1

    def toon_teller(self):
        print(f"Huidige waarde van TELLER: {TELLER}")

# Test de class
machine = TelMachine()

# Verhoog de teller en toon de waarde
machine.verhoog()
machine.toon_teller()  

machine.verhoog()
machine.toon_teller()  

machine.verhoog()
machine.toon_teller()  
```
