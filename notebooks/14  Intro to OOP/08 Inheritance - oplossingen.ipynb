### Oefening 1: 
Maak een klasse genaamd `Fruit` en twee subklassen `Appel` en `Peer`. Gebruik een class method om een nieuw stuk fruit te maken. Implementeer de volgende eigenschappen:

- **Attributen**
  - `naam`: De naam van het fruit.
  - `kleur`: De kleur van het fruit.
- **Class Method**
  - `maak_fruit(soort, kleur)`: Maak een nieuw stuk fruit op basis van de soort (appel of peer).

````python
class Fruit:
    def __init__(self, naam, kleur):
        self.naam = naam
        self.kleur = kleur
    
    @classmethod
    def maak_fruit(cls, soort, kleur):
        if soort.lower() == "appel":
            return Appel(kleur)
        else:
            if soort.lower() == "peer":
                return Peer(kleur)

class Appel(Fruit):
    def __init__(self, kleur):
        super().__init__("Appel", kleur)

class Peer(Fruit):
    def __init__(self, kleur):
        super().__init__("Peer", kleur)

a = Fruit.maak_fruit("appel", "rood")
print(a.naam, a.kleur)  # Uitvoer: Appel rood
````

### Oefening 2: 
Maak een klasse `Voertuig` en een afgeleide klasse `Auto`:
   - `Voertuig` moet attributen bevatten zoals merk en bouwjaar.
   - `Auto` moet extra attributen hebben zoals `model` en een methode `start()` die aangeeft dat de auto start.

````python
class Voertuig:
    def __init__(self, merk, bouwjaar):
        self.merk = merk
        self.bouwjaar = bouwjaar

    def __str__(self):
        return f"{self.merk} ({self.bouwjaar})"

class Auto(Voertuig):
    def __init__(self, merk, bouwjaar, model):
        super().__init__(merk, bouwjaar)
        self.model = model

    def start(self):
        return f"De {self.merk} {self.model} start."

    def __str__(self):
        return f"{self.merk} {self.model} ({self.bouwjaar})"

# Voorbeeld van het gebruik:
auto = Auto("Toyota", 2020, "Corolla")
print(auto)
print(auto.start())
````

### Oefening 3:

Maak een `Persoon` klasse en een afgeleide klasse `Werknemer`:
   - `Persoon` moet attributen hebben zoals naam en leeftijd.
   - `Werknemer` moet extra attributen hebben zoals `functie` en `salaris`.
   - Voeg een methode toe om de werknemerinformatie af te drukken.

````python
class Persoon:
    def __init__(self, naam, leeftijd):
        self.naam = naam
        self.leeftijd = leeftijd

    def __str__(self):
        return f"Naam: {self.naam}, Leeftijd: {self.leeftijd}"

class Werknemer(Persoon):
    def __init__(self, naam, leeftijd, functie, salaris):
        super().__init__(naam, leeftijd)
        self.functie = functie
        self.salaris = salaris

    def print_werknemer_info(self):
        print(f"{self}, Functie: {self.functie}, Salaris: {self.salaris}")

# Voorbeeldgebruik
werknemer = Werknemer("Jan", 35, "Softwareontwikkelaar", 4000)
werknemer.print_werknemer_info()
````

### Oefening 6:
Maak een hiërarchie van klassen: `Apparaat` -> `Computer` -> `Laptop`**:
   - `Apparaat` moet een attribuut hebben voor `merk`.
   - `Computer` moet een extra attribuut hebben voor `processor`.
   - `Laptop` moet een extra attribuut hebben voor `gewicht`.
   - Voeg methoden toe in elke klasse om de details weer te geven.

````python
class Apparaat:
    def __init__(self, merk):
        self.merk = merk

    def toon_details(self):
        return f"Merk: {self.merk}"

class Computer(Apparaat):
    def __init__(self, merk, processor):
        super().__init__(merk)
        self.processor = processor

    def toon_details(self):
        return f"{super().toon_details()}, Processor: {self.processor}"

class Laptop(Computer):
    def __init__(self, merk, processor, gewicht):
        super().__init__(merk, processor)
        self.gewicht = gewicht

    def toon_details(self):
        return f"{super().toon_details()}, Gewicht: {self.gewicht} kg"

# Voorbeeldgebruik
laptop = Laptop("Dell", "Intel i7", 1.5)
print(laptop.toon_details())
````

### Oefening 8: Python Multiple Inheritance  


Maak een Python-programma dat verschillende soorten voertuigen modelleert met behulp van **multiple inheritance**. De opdracht is als volgt:

---

#### **Situatie**
Je bent gevraagd om een applicatie te maken om eigenschappen van voertuigen bij te houden. Er zijn twee basiscategorieën:
1. **LandVoertuig**: Voertuigen die op land rijden, zoals auto's en fietsen.
2. **WaterVoertuig**: Voertuigen die op water bewegen, zoals boten en jetski's.

Sommige voertuigen kunnen echter op beide terreinen functioneren, zoals een amfibievoertuig.

---

#### **Opdracht**
1. Maak de volgende klassen:
   - **LandVoertuig**:
     - Attributen:
       - `max_snelheid_land` (maximale snelheid op land)
     - Methoden:
       - `rijden()`: Print een bericht zoals `"Ik rij met X km/u op land."`

   - **WaterVoertuig**:
     - Attributen:
       - `max_snelheid_water` (maximale snelheid op water)
     - Methoden:
       - `varen()`: Print een bericht zoals `"Ik vaar met Y km/u op water."`

   - **AmfibieVoertuig** (erft van zowel LandVoertuig als WaterVoertuig):
     - Methode:
       - `toestand()`: Print een bericht zoals `"Ik kan rijden en varen."`

2. Maak objecten voor elk type voertuig:
   - Een **LandVoertuig** met een maximale snelheid van 120 km/u.
   - Een **WaterVoertuig** met een maximale snelheid van 50 km/u.
   - Een **AmfibieVoertuig** dat kan rijden met 100 km/u en varen met 40 km/u.

3. Demonstreer het gebruik van alle methoden voor elk object.


**Extra uitdaging**:
1. Voeg een methode `status()` toe aan de `AmfibieVoertuig`-klasse die beide snelheden print, zoals:
   ```
   Ik kan rijden met 100 km/u en varen met 40 km/u.
   ```
2. Voeg een controle toe om ervoor te zorgen dat het voertuig niet beide acties tegelijkertijd kan uitvoeren.

````python
class LandVoertuig:
    def __init__(self, max_snelheid_land):
        self.max_snelheid_land = max_snelheid_land

    def rijden(self):
        print(f"Ik rij met {self.max_snelheid_land} km/u op land.")

class WaterVoertuig:
    def __init__(self, max_snelheid_water):
        self.max_snelheid_water = max_snelheid_water

    def varen(self):
        print(f"Ik vaar met {self.max_snelheid_water} km/u op water.")

class AmfibieVoertuig(LandVoertuig, WaterVoertuig):
    def __init__(self, max_snelheid_land, max_snelheid_water):
        LandVoertuig.__init__(self, max_snelheid_land)
        WaterVoertuig.__init__(self, max_snelheid_water)

    def toestand(self):
        print("Ik kan rijden en varen.")

# Objecten aanmaken
auto = LandVoertuig(120)
boot = WaterVoertuig(50)
amfibie = AmfibieVoertuig(100, 40)

# Methoden demonstreren
auto.rijden()
boot.varen()
amfibie.rijden()
amfibie.varen()
amfibie.toestand()
````

````python
# Extra uitdaging:

class LandVoertuig:
    def __init__(self, max_snelheid_land):
        self.max_snelheid_land = max_snelheid_land
        self.is_aan_het_rijden = False

    def rijden(self):
        if not self.is_aan_het_rijden:
            self.is_aan_het_rijden = True
            print(f"Ik rij met {self.max_snelheid_land} km/u op land.")
        else:
            print("Ik ben al aan het rijden.")

    def stoppen(self):
        if self.is_aan_het_rijden:
            self.is_aan_het_rijden = False
            print("Ik ben gestopt met rijden.")

class WaterVoertuig:
    def __init__(self, max_snelheid_water):
        self.max_snelheid_water = max_snelheid_water
        self.is_aan_het_varen = False

    def varen(self):
        if not self.is_aan_het_varen:
            self.is_aan_het_varen = True
            print(f"Ik vaar met {self.max_snelheid_water} km/u op water.")
        else:
            print("Ik ben al aan het varen.")

    def stoppen(self):
        if self.is_aan_het_varen:
            self.is_aan_het_varen = False
            print("Ik ben gestopt met varen.")

class AmfibieVoertuig(LandVoertuig, WaterVoertuig):
    def __init__(self, max_snelheid_land, max_snelheid_water):
        LandVoertuig.__init__(self, max_snelheid_land)
        WaterVoertuig.__init__(self, max_snelheid_water)

    def toestand(self):
        print("Ik kan rijden en varen.")

    def status(self):
        print(f"Ik kan rijden met {self.max_snelheid_land} km/u en varen met {self.max_snelheid_water} km/u.")

    def rijden(self):
        if not self.is_aan_het_varen:
            super().rijden()
        else:
            print("Ik kan niet rijden terwijl ik aan het varen ben.")

    def varen(self):
        if not self.is_aan_het_rijden:
            super().varen()
        else:
            print("Ik kan niet varen terwijl ik aan het rijden ben.")

# Objecten aanmaken
auto = LandVoertuig(120)
boot = WaterVoertuig(50)
amfibie = AmfibieVoertuig(100, 40)

# Methoden demonstreren
auto.rijden()
auto.rijden()
auto.stoppen()
boot.varen()
boot.stoppen()
amfibie.rijden()
amfibie.stoppen()
amfibie.varen()
amfibie.stoppen()
amfibie.status()
amfibie.toestand()
````
