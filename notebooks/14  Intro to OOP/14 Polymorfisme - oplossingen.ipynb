### Oefening 1: Dieren Simuleren
   - Maak vier dieren aan: een leeuw, giraf, kat, en hond.
   - Elk dier moet een geluid kunnen maken, een lengte hebben (klein, gemiddeld, groot, zeer groot) en een loopsnelheid (uitgedrukt in seconde per meter).
   - Plaats al deze dieren in een lijst en druk voor elk dier het type, de lengte, en de loopsnelheid af.

```python
class Dier:
    def __init__(self, naam, lengte, snelheid):
        self.naam = naam
        self.lengte = lengte
        self.snelheid = snelheid

    def maak_geluid(self):
        pass

    def beschrijf(self):
        return f"{self.naam}: {self.lengte}, snelheid: {self.snelheid} sec/meter"

class Leeuw(Dier):
    def maak_geluid(self):
        return "Roar!"

class Giraf(Dier):
    def maak_geluid(self):
        return "Hmm!"

class Kat(Dier):
    def maak_geluid(self):
        return "Miauw!"

class Hond(Dier):
    def maak_geluid(self):
        return "Woef!"

dieren = [
    Leeuw("Leeuw", "groot", 3),
    Giraf("Giraf", "zeer groot", 5),
    Kat("Kat", "klein", 1),
    Hond("Hond", "gemiddeld", 2)
]

for dier in dieren:
    print(dier.beschrijf())
```

### Oefening 2: Berekeningen van Figuren
   - Maak een programma om de omtrek en oppervlakte te berekenen van de volgende figuren: vierkant, rechthoek, cirkel, driehoek.
   - Vraag de gebruiker welk type hij/zij wil berekenen en vraag de benodigde maten op.
   - Gebruik polymorfisme om de methoden voor elke figuur te implementeren en druk het resultaat af.

```python
import math

class Figuur:
    def omtrek(self):
        pass

    def oppervlakte(self):
        pass

class Vierkant(Figuur):
    def __init__(self, zijde):
        self.zijde = zijde

    def omtrek(self):
        return 4 * self.zijde

    def oppervlakte(self):
        return self.zijde ** 2

class Rechthoek(Figuur):
    def __init__(self, lengte, breedte):
        self.lengte = lengte
        self.breedte = breedte

    def omtrek(self):
        return 2 * (self.lengte + self.breedte)

    def oppervlakte(self):
        return self.lengte * self.breedte

class Cirkel(Figuur):
    def __init__(self, straal):
        self.straal = straal

    def omtrek(self):
        return 2 * math.pi * self.straal

    def oppervlakte(self):
        return math.pi * (self.straal ** 2)

class Driehoek(Figuur):
    def __init__(self, basis, hoogte):
        self.basis = basis
        self.hoogte = hoogte

    def omtrek(self):
        return "Niet genoeg informatie om de omtrek te berekenen."

    def oppervlakte(self):
        return 0.5 * self.basis * self.hoogte

# Voorbeeldgebruik
figuur = Vierkant(4)
print(f"Omtrek: {figuur.omtrek()}")
print(f"Oppervlakte: {figuur.oppervlakte()}")
```
