### **Oefening 1: Custom Exception**
Maak een custom exception `OngeldigeTransactie` die wordt gebruikt in een class `Rekening`. De class moet een methode `opnemen` hebben om geld op te nemen. Controleer de volgende voorwaarden:
1. Het opnamebedrag mag niet groter zijn dan het saldo.
2. Het opnamebedrag moet positief zijn.

**Test:**  
- Probeer geld op te nemen met een bedrag groter dan het saldo en een negatief bedrag. Zorg dat de juiste foutmelding wordt weergegeven.


```python
# oplossing



class Rekening:
    def __init__(self, saldo):
        self.saldo = saldo

    def opnemen(self, bedrag):
        # Controleer of het bedrag positief is
        if bedrag <= 0:
            raise OngeldigeTransactie("Het opnamebedrag moet positief zijn.")
        
        # Controleer of er voldoende saldo is
        if bedrag > self.saldo:
            raise OngeldigeTransactie("Onvoldoende saldo om dit bedrag op te nemen.")
        
        # Trek het bedrag af van het saldo
        self.saldo -= bedrag
        return self.saldo

class OngeldigeTransactie(Exception):
    def __init__(self, bericht):
        super().__init__(bericht)

# Test de class
rekening = Rekening(1000)

# Test 1: Geldige opname
try:
    nieuw_saldo = rekening.opnemen(200)
    print(f"Opname gelukt. Nieuw saldo: {nieuw_saldo}")
except OngeldigeTransactie as e:
    print(f"Fout: {e}")

# Test 2: Ongeldige opname - bedrag groter dan saldo
try:
    nieuw_saldo = rekening.opnemen(1500)
    print(f"Opname gelukt. Nieuw saldo: {nieuw_saldo}")
except OngeldigeTransactie as e:
    print(f"Fout: {e}")  

# Test 3: Ongeldige opname - negatief bedrag
try:
    nieuw_saldo = rekening.opnemen(-100)
    print(f"Opname gelukt. Nieuw saldo: {nieuw_saldo}")
except OngeldigeTransactie as e:
    print(f"Fout: {e}")  
```

### **Oefening 2: Meerdere except-blokken**
Maak een class `Rekenmachine` met een methode `deel`. Deze methode accepteert twee getallen en retourneert het quotiënt. Controleer de volgende fouten:
1. Als de deler 0 is, geef een `ZeroDivisionError`.
2. Als een van de invoerwaarden geen getal is, geef een `TypeError`.

**Test:**  
- Roep de methode aan met een geldige en ongeldige deler en zorg dat elke fout correct wordt afgehandeld.


```python
# oplossing

class Rekenmachine:
    def deel(self, a, b):
        # Controleer of beide inputs getallen zijn
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise TypeError("Beide invoerwaarden moeten getallen zijn.")
        
        # Controleer of de deler 0 is
        if b == 0:
            raise ZeroDivisionError("Delen door nul is niet toegestaan.")
        
        # Retourneer het quotiënt
        return a / b

# Test de class
rekenmachine = Rekenmachine()

# Geldige invoer
try:
    resultaat = rekenmachine.deel(10, 2)
    print(f"Resultaat: {resultaat}")  # Output: 5.0
except Exception as e:
    print(f"Fout: {e}")

# Ongeldige invoer: delen door 0
try:
    resultaat = rekenmachine.deel(10, 0)
    print(f"Resultaat: {resultaat}")
except Exception as e:
    print(f"Fout: {e}")  

# Ongeldige invoer: niet-numerieke waarde
try:
    resultaat = rekenmachine.deel(10, "twee")
    print(f"Resultaat: {resultaat}")
except Exception as e:
    print(f"Fout: {e}")  
```

### **Oefening 3: `try-except` in een methode**
Maak een class `BestandManager` die een methode `lees_bestand` heeft om de inhoud van een tekstbestand te lezen. Als het bestand niet bestaat, moet een `FileNotFoundError` worden afgehandeld in de methode.

**Test:**  
- Probeer een niet-bestaand bestand te openen en geef een foutmelding zoals: *"Bestand niet gevonden"*.

```python
# oplossing

class BestandManager:
    def lees_bestand(self, bestandsnaam):
        try:
            with open(bestandsnaam, 'r') as bestand:
                inhoud = bestand.read()
                return inhoud
        except FileNotFoundError:
            return "Bestand niet gevonden."

# Test de class
manager = BestandManager()

# Test 1: Lees een bestaand bestand
try:
    inhoud = manager.lees_bestand("bestaand_bestand.txt")
    print("Inhoud van het bestand:")
    print(inhoud)
except Exception as e:
    print(f"Onverwachte fout: {e}")

# Test 2: Probeer een niet-bestaand bestand te openen
try:
    inhoud = manager.lees_bestand("niet_bestaand_bestand.txt")
    print("Inhoud van het bestand:")
    print(inhoud)  
except Exception as e:
    print(f"Onverwachte fout: {e}")
```

### **Oefening 4: Hergebruik van fouten in een subklasse**
Maak een basisclass `BankRekening` met foutafhandeling in de methode `stort` en `opname`. Maak een subklasse `ZakelijkeRekening` die extra regels toevoegt:
1. Het saldo mag niet onder een bepaald minimumbedrag komen (bijv. 500 euro).
2. Implementeer aanvullende foutmeldingen voor deze regels.

**Test:**  
- Test de subklasse met bedragen die de regels overtreden.

```python
# oplossing
class BankRekening:
    def __init__(self, saldo):
        self.saldo = saldo

    def stort(self, bedrag):
        if bedrag <= 0:
            raise ValueError("Het stortingsbedrag moet positief zijn.")
        self.saldo += bedrag
        return self.saldo

    def opname(self, bedrag):
        if bedrag <= 0:
            raise ValueError("Het opnamebedrag moet positief zijn.")
        if bedrag > self.saldo:
            raise ValueError("Onvoldoende saldo om dit bedrag op te nemen.")
        self.saldo -= bedrag
        return self.saldo

class ZakelijkeRekening(BankRekening):
    MINIMUM_SALDO = 500

    def opname(self, bedrag):
        if bedrag <= 0:
            raise ValueError("Het opnamebedrag moet positief zijn.")
        if self.saldo - bedrag < self.MINIMUM_SALDO:
            raise ValueError(f"Het saldo mag niet onder het minimum van {self.MINIMUM_SALDO} euro komen.")
        return super().opname(bedrag)

# Test de classes
try:
    # Basisclass test
    rekening = BankRekening(1000)
    print("Begin saldo:", rekening.saldo)
    rekening.stort(200)
    print("Na storting van 200 euro:", rekening.saldo)
    rekening.opname(300)
    print("Na opname van 300 euro:", rekening.saldo)
    
    # Ongeldige storting
    try:
        rekening.stort(-50)
    except ValueError as e:
        print(f"Fout: {e}")  

    # Ongeldige opname
    try:
        rekening.opname(1500)
    except ValueError as e:
        print(f"Fout: {e}")  

    # Zakelijke rekening test
    zakelijke_rekening = ZakelijkeRekening(1000)
    print("Begin saldo zakelijke rekening:", zakelijke_rekening.saldo)
    zakelijke_rekening.opname(400)
    print("Na opname van 400 euro:", zakelijke_rekening.saldo)
    
    # Ongeldige opname - onder minimumsaldo
    try:
        zakelijke_rekening.opname(200)
    except ValueError as e:
        print(f"Fout: {e}")  
except Exception as e:
    print(f"Onverwachte fout: {e}")
```

### **Oefening 5: Foutenrapportage met `__str__`**
Maak een class `FoutLogger` die fouten registreert. Voeg een methode `log_fout` toe om fouten op te slaan. Implementeer `__str__` om een leesbare lijst van fouten weer te geven.

**Test:**  
- Registreer meerdere fouten en print de `FoutLogger` om te zien hoe fouten worden weergegeven.

```python
# oplossing
class FoutLogger:
    def __init__(self):
        self.fouten = []  # Lijst om fouten op te slaan

    def log_fout(self, fout):
        self.fouten.append(fout)  # Voeg een fout toe aan de lijst

    def __str__(self):
        if not self.fouten:
            return "Geen fouten geregistreerd."
        return "\n".join(f"{i + 1}. {fout}" for i, fout in enumerate(self.fouten))

# Test de class
logger = FoutLogger()

# Registreer fouten
logger.log_fout("Bestand niet gevonden.")
logger.log_fout("Ongeldige invoer.")
logger.log_fout("Verbinding verbroken.")

# Print de logger
print(logger)

# Test met een lege logger
lege_logger = FoutLogger()
print(lege_logger)  
```

### **Oefening 6: Bankrekening met error handling en logging**  

Pas de code van oefening 4 aan om logging te gebruiken in plaats van errors via print statements weer te geven.  

Het doel is om **logging** toe te voegen om foutmeldingen en transacties beter te beheren. In plaats van `print()` te gebruiken, wordt de **`logging` module** gebruikt om alle transacties en fouten vast te leggen.  

#### Vereisten

1. **Logging toevoegen aan de klassen**:
   - **BankRekening**:
     - Voeg een **`logging.info`**-bericht toe voor succesvolle stortingen en opnames.
     - Voeg een **`logging.error`**-bericht toe voor fouten, zoals een negatief bedrag storten of opnemen, of een poging om meer op te nemen dan het saldo.
   - **ZakelijkeRekening**:
     - Log een foutmelding met **`logging.error`** als een opname het saldo onder het minimumsaldo brengt.

2. **Logging configureren**:
   - Stel de `logging` module in om berichten op niveau **`INFO`** en hoger te loggen.
   - Gebruik het format: `%(asctime)s - %(levelname)s - %(message)s`.
   - Zorg ervoor dat de logs worden opgeslagen in een bestand genaamd `bank_transacties.log`.

3. **Gebruik duidelijke foutmeldingen**:
   - Werp een `ValueError` met een duidelijke foutmelding wanneer een ongeldige transactie wordt geprobeerd (bijvoorbeeld een negatief bedrag of onvoldoende saldo).

4. **Test de klassen**:
   - Implementeer de volgende tests:
     - Maak een `BankRekening` en voer een storting en opname uit.
     - Probeer een negatief bedrag te storten en log de fout.
     - Probeer een opname te doen die het saldo overschrijdt en log de fout.
     - Maak een `ZakelijkeRekening` en voer een geldige opname uit.
     - Probeer een opname te doen die het saldo onder het minimum brengt en log de fout.


#### Verwachte Uitbreidingen
- **Logging**: 
  - Gebruik de `logging` module in plaats van `print()`.
  - Log alle transacties en fouten met een duidelijke melding.
- **Configuratie**:
  - Stel een logbestand in (`bank_transacties.log`) waarin transacties en fouten worden vastgelegd.

#### Evaluatiecriteria
1. **Correct gebruik van logging**:
   - Succesvolle transacties worden gelogd met `INFO`.
   - Fouten worden gelogd met `ERROR`.
2. **Testcases**:
   - Alle gevraagde tests werken correct en logs bevatten relevante informatie.
3. **Duidelijke foutmeldingen**:
   - Elke fout geeft een nuttige en begrijpelijke foutmelding in de logs.

Na afloop van de oefening moet je een bestand `bank_transacties.log` hebben waarin alle transacties en fouten duidelijk zijn vastgelegd. Gebruik dit bestand om je oplossing te controleren!

```python
import logging

# Logging configureren
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filename="bank_transacties.log",  # Optioneel: log in een bestand
    filemode="a"  # Append modus
)

class BankRekening:
    def __init__(self, saldo):
        self.saldo = saldo
        logging.info(f"Bankrekening aangemaakt met startsaldo: {self.saldo} euro.")

    def stort(self, bedrag):
        if bedrag <= 0:
            logging.error(f"Probeer een negatief of nul bedrag te storten: {bedrag}")
            raise ValueError("Het stortingsbedrag moet positief zijn.")
        self.saldo += bedrag
        logging.info(f"Storting van {bedrag} euro uitgevoerd. Nieuw saldo: {self.saldo} euro.")
        return self.saldo

    def opname(self, bedrag):
        if bedrag <= 0:
            logging.error(f"Probeer een negatief of nul bedrag op te nemen: {bedrag}")
            raise ValueError("Het opnamebedrag moet positief zijn.")
        if bedrag > self.saldo:
            logging.error(f"Opname van {bedrag} euro geweigerd. Onvoldoende saldo: {self.saldo} euro.")
            raise ValueError("Onvoldoende saldo om dit bedrag op te nemen.")
        self.saldo -= bedrag
        logging.info(f"Opname van {bedrag} euro uitgevoerd. Nieuw saldo: {self.saldo} euro.")
        return self.saldo

class ZakelijkeRekening(BankRekening):
    MINIMUM_SALDO = 500

    def opname(self, bedrag):
        if bedrag <= 0:
            logging.error(f"Probeer een negatief of nul bedrag op te nemen: {bedrag}")
            raise ValueError("Het opnamebedrag moet positief zijn.")
        if self.saldo - bedrag < self.MINIMUM_SALDO:
            logging.error(
                f"Opname van {bedrag} euro geweigerd. "
                f"Saldo na opname zou onder het minimum van {self.MINIMUM_SALDO} euro komen."
            )
            raise ValueError(f"Het saldo mag niet onder het minimum van {self.MINIMUM_SALDO} euro komen.")
        return super().opname(bedrag)
```

```python
# Test de classes
try:
    # Basisclass test
    rekening = BankRekening(1000)
    logging.info(f"Start saldo: {rekening.saldo}")
    rekening.stort(200)
    logging.info(f"Saldo na storting: {rekening.saldo}")
    rekening.opname(300)
    logging.info(f"Saldo na opname: {rekening.saldo}")
    
    # Ongeldige storting
    try:
        rekening.stort(-50)
    except ValueError as e:
        logging.error(f"Fout tijdens storting: {e}")

    # Ongeldige opname
    try:
        rekening.opname(1500)
    except ValueError as e:
        logging.error(f"Fout tijdens opname: {e}")

    # Zakelijke rekening test
    zakelijke_rekening = ZakelijkeRekening(1000)
    logging.info(f"Start saldo zakelijke rekening: {zakelijke_rekening.saldo}")
    zakelijke_rekening.opname(400)
    logging.info(f"Saldo na opname: {zakelijke_rekening.saldo}")
    
    # Ongeldige opname - onder minimumsaldo
    try:
        zakelijke_rekening.opname(200)
    except ValueError as e:
        logging.error(f"Fout tijdens opname zakelijke rekening: {e}")
except Exception as e:
    logging.critical(f"Onverwachte fout: {e}")
```
