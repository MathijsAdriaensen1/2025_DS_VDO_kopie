### Oefening 1: Toegang tot Items  

Maak een class Database die een interne dictionary gebruikt voor opslag. Implementeer:  

- __getitem__: Haal een waarde op met een sleutel.
- __setitem__: Voeg een nieuwe sleutel-waarde-paar toe.
- __delitem__: Verwijder een sleutel-waarde-paar.

**voorbeeld gebruik:**  
```python
db = Database()  
db["gebruikersnaam"] = "admin"  
print(db["gebruikersnaam"])  # Output: admin  
del db["gebruikersnaam"]
```

````python
# oplossing
class Database:
    def __init__(self):
        self.storage = {}  # Interne opslag als dictionary

    def __getitem__(self, key):
        return self.storage.get(key, "Sleutel niet gevonden")  # Ophalen van waarde

    def __setitem__(self, key, value):
        self.storage[key] = value  # Instellen van waarde

    def __delitem__(self, key):
        if key in self.storage:
            del self.storage[key]  # Verwijderen van sleutel
        else:
            print(f"Sleutel '{key}' bestaat niet.")

# Voorbeeldgebruik
db = Database()
db["gebruikersnaam"] = "admin"
print(db["gebruikersnaam"])  
del db["gebruikersnaam"]
print(db["gebruikersnaam"])  
````

### Oefening 2: Aangepaste Functie-objecten  

Maak een class Vermenigvuldiger die een factor accepteert bij initialisatie. Implementeer:  

- __call__: Maak de class "aanroepbaar" zodat het een waarde vermenigvuldigt met de factor.

**voorbeeld gebruik:**
```python
vermenigvuldig = Vermenigvuldiger(3)  
print(vermenigvuldig(10))  # Output: 30
```

````python
# oplossing
class Vermenigvuldiger:
    def __init__(self, factor):
        self.factor = factor  # De vermenigvuldigingsfactor

    def __call__(self, waarde):
        return waarde * self.factor  # Voer de vermenigvuldiging uit

# Voorbeeldgebruik
vermenigvuldig = Vermenigvuldiger(3)
print(vermenigvuldig(10))  
print(vermenigvuldig(7))  
````

### Oefening 3: Context Manager  

Maak een class Timer die een context manager implementeert om de tijd te meten tussen het begin en einde van een blok code. Gebruik:  

- __enter__: Start de timer.
- __exit__: Stop de timer en toon de verstreken tijd.

**voorbeeld gebruik:**  
```python
with Timer():
    for i in range(1000000):
        pass
```

````python
# oplossing
import time

class Timer:
    def __enter__(self):
        self.start = time.time()  # Starttijd
        print("Timer gestart...")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        self.end = time.time()  # Eindtijd
        print(f"Verstreken tijd: {self.end - self.start:.2f} seconden")

# Voorbeeldgebruik
with Timer():
    for i in range(1000000):
        pass
````

### Oefening 4: __contains__  

Maak een class Bibliotheek die een lijst van boeken accepteert. Implementeer:

__contains__: Controleer of een bepaald boek in de bibliotheek zit.

**voorbeeld gebruik:**
```python
bieb = Bibliotheek(["1984", "Brave New World"])
print("1984" in bieb)  # Output: True
```

````python
# oplossing
class Bibliotheek:
    def __init__(self, boeken):
        self.boeken = boeken  # Lijst van boeken

    def __contains__(self, boek):
        return boek in self.boeken  # Controleer of boek in de lijst zit

# Voorbeeldgebruik
bieb = Bibliotheek(["1984", "Brave New World", "Animal Farm"])
print("1984" in bieb)        
print("Harry Potter" in bieb)  
````
