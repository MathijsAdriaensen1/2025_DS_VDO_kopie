## Oefening: Regularisatie in Lineaire Regressie**

### **Doel**
In deze oefening ga je werken met **Ridge-, Lasso- en Elastic Net-regressie**. Je leert hoe deze methoden helpen bij het **voorkomen van overfitting**, hoe je de **belangrijkste parameters** selecteert en hoe je de invloed van de features **visualiseert**.


### **Stap 1: Importeren van de dataset**
Gebruik de **Diabetes dataset** uit `sklearn.datasets` (from sklearn.datasets import load_diabetes)

1. **Importeer de nodige modules:**
   - `pandas`, `numpy`, `matplotlib.pyplot`
   - `train_test_split`, `StandardScaler`
   - `LinearRegression`, `Ridge`, `Lasso`, `ElasticNet`
   - `r2_score`, `cross_val_score`, `GridSearchCV`

2. **Laad de dataset** en splits de data in **features (X)** en **target (y)**.
3. **Schaal de data** met `StandardScaler`.
4. **Splits de dataset** in een **trainingsset** en een **testset** (80/20-verhouding).


### **Stap 2: Lineaire Regressie zonder Regularisatie**
1. **Train een lineair regressiemodel** met `LinearRegression()`.
2. **Bereken de R²-score** voor zowel de **trainingsset** als de **testset**.
3. **Visualiseer de coëfficiënten** in een horizontale barplot.
4. **Beantwoord:** 
   - Hoe goed presteert het model op de testset?
   - Is er een risico op overfitting?


### **Stap 3: Ridge Regressie**
1. **Train een Ridge-model** (`Ridge()`).
2. **Test met verschillende waarden van alpha** (bijv. 0.01, 0.1, 1, 10, 100).
3. **Gebruik `GridSearchCV`** om de **beste alpha-waarde** te bepalen.
4. **Bereken de R²-score** en vergelijk met lineaire regressie.
5. **Visualiseer de coëfficiënten**.
6. **Beantwoord:**
   - Hoe verandert de prestatie t.o.v. lineaire regressie?
   - Wat gebeurt er met de coëfficiënten bij hogere waarden van alpha?


### **Stap 4: Lasso Regressie**
1. **Train een Lasso-model** (`Lasso()`).
2. **Test met verschillende waarden van alpha**.
3. **Gebruik `GridSearchCV`** om de **beste alpha-waarde** te bepalen.
4. **Bereken de R²-score** en vergelijk met Ridge.
5. **Visualiseer de coëfficiënten** en kijk welke coëfficiënten op nul staan.
6. **Beantwoord:**
   - Wat gebeurt er met minder belangrijke features?
   - Wat is het verschil tussen Ridge en Lasso?


### **Stap 5: Elastic Net Regressie**
1. **Train een Elastic Net-model** (`ElasticNet()`).
2. **Test met verschillende combinaties van alpha en l1_ratio**.
3. **Gebruik `GridSearchCV`** om de **beste combinatie van hyperparameters** te vinden.
4. **Bereken de R²-score** en vergelijk met Ridge en Lasso.
5. **Visualiseer de coëfficiënten**.
6. **Beantwoord:**
   - Wanneer presteert Elastic Net beter dan Ridge of Lasso?
   - Wat is het effect van de l1_ratio?
