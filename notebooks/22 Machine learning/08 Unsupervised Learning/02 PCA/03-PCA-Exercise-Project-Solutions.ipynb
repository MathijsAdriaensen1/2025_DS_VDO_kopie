## Schalen van de Data

**TAAK: Gebruik Scikit-Learn om het feature dataframe te schalen.**

```python
#CODE HERE
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaled_pixels = scaler.fit_transform(pixels)
```

```python
scaled_pixels
```

## PCA

**TAAK: Voer PCA uit op de geschaalde pixel data set met 2 components.**

```python
from sklearn.decomposition import PCA
```

```python
pca_model = PCA(n_components=2)
```

```python
pca_pixels = pca_model.fit_transform(scaled_pixels)
```

**TAAK: Hoeveel variantie is er verklaard met de 2 principal components?**

```python
#CODE HERE
```

```python
np.sum(pca_model.explained_variance_ratio_)
```

**TAAK: Maak een scatterplot van de cijfers in de 2-dimensionale PCA-ruimte, waarbij kleur/label op basis van de originele kolom number_label in de oorspronkelijke dataset is.**

```python
#CODE HERE
```

```python
plt.figure(figsize=(10,6),dpi=150)
labels = digits['number_label'].values
sns.scatterplot(x=pca_pixels[:,0],y=pca_pixels[:,1],hue=labels,palette='Set1')
plt.legend(loc=(1.05,0))
```
