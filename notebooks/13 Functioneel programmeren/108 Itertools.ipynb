### Introductie tot de itertools module in Python

De `itertools` module in Python biedt een verzameling krachtige tools voor het efficiënt werken met iteraties. Het stelt je in staat om complexe iteraties te maken zonder dat je zelf uitgebreide loops hoeft te schrijven. 

`itertools` bevat functies die je kunt zien als "oneindige gereedschappen" om iteraties te manipuleren en combineren.

Laten we een eenvoudige introductie geven tot enkele veelgebruikte functies:

### 1. **Iterators maken van een `range`-object**
```python
r = range(1, 10)
i = iter(r)  # Maakt een iterator

print(next(i))  # 1
print(next(i))  # 2
```
**Uitleg:**  
- `iter()` maakt een expliciete iterator van het `range`-object.  
- `next()` haalt het volgende element op.  
**Waarom handig?**  
Met iterators kun je elementen stap voor stap doorlopen zonder het hele object in het geheugen te laden.

### 2. **`next()` gebruiken om de iterator te doorlopen**
```python
r = range(1, 10)
i = iter(r)

print(next(i))  # 1
print(next(i))  # 2
print(next(i))  # 3
```
**Uitleg:**  
- `next()` haalt het volgende element van de iterator op.  
**Waarom handig?**  
Handig bij gecontroleerde iteratie, bijvoorbeeld wanneer je specifieke elementen wilt ophalen.

### 3. **`StopIteration`-fout bij het doorlopen van de iterator**
```python
try:
    for _ in range(10):
        print(next(i))
except StopIteration as e:
    print("StopIteration error caught")
```
**Uitleg:**  
- Een `StopIteration`-fout treedt op wanneer er geen elementen meer zijn in de iterator.  
**Waarom handig?**  
Dit illustreert het gedrag van iterators en hoe je een eindconditie kunt detecteren.

### 4. **`itertools.count` - Oneindige iterator**
```python
from itertools import count
count_example(10, 5)  # (start, step)
```
**Uitleg:**  
- `count()` maakt een oneindige reeks getallen met een specifieke stapgrootte.  
**Waarom handig?**  
Perfect voor toepassingen waar je doorlopend moet tellen, zoals ID-generatie of numerieke tests.

### 5. **`itertools.repeat`**
```python
from itertools import repeat
repeat_example("hello", 10)
```
**Uitleg:**  
- `repeat()` herhaalt een element een vast aantal keer of oneindig.  
**Waarom handig?**  
Nuttig voor testdoeleinden of bij het aanmaken van constante waardes in loops.

### 6. **`itertools.cycle`**
```python
from itertools import cycle
cycle_example("ABCDEF")
```
**Uitleg:**  
- `cycle()` doorloopt de elementen oneindig in een lus.  
**Waarom handig?**  
Ideaal voor herhalende patronen zoals animaties, simulaties, of doorlopende menu’s.

### 7. **`itertools.accumulate`**
```python
from itertools import accumulate
accumulate_example([1, 2, 3, 4, 5])
```
**Uitleg:**  
- `accumulate()` geeft een lopende som (of andere cumulatieve berekeningen).  
**Waarom handig?**  
Handig voor berekeningen zoals voortschrijdende sommen, gemiddelden of andere aggregaties.

Je kan ook een functie meegeven, die dan een andere bewerking op de elementen uitvoert (in plaats van standaard optellen):  

itertools.accumulate(iterable, func=operator.add)

````python
from itertools import accumulate

result = list(accumulate([1, 2, 3, 4, 5], lambda x, y: x*y))
print(result)
````

### 8. **`itertools.chain`**
```python
from itertools import chain
chain_example("ABC", "DEF")
```
**Uitleg:**  
- `chain()` combineert meerdere iterables achter elkaar.  
**Waarom handig?**  
Handig om lijsten, tuplereeksen of andere iterables efficiënt samen te voegen.

### 9. **`itertools.chain.from_iterable`**
```python
from itertools import chain
chain.from_iterable([[1, 2], [3, 4], [5, 6]])
```
**Uitleg:**  
- `chain.from_iterable()` neemt een iterable van iterables en voegt ze samen.  
**Waarom handig?**  
Nuttig bij het "plat maken" van lijsten met lijsten.

````python
from itertools import chain
list(chain([1, 2], [3, 4], [5, 6]))
````

````python
list(chain([[1, 2], [3, 4], [5, 6]]))
````

````python
list(chain.from_iterable([[1, 2], [3, 4], [5, 6]]))
````

### 10. **`itertools.compress`**
```python
from itertools import compress
compress_example(["a", "b", "c"], [True, False, True])
```
**Uitleg:**  
- `compress()` filtert elementen op basis van een selector (True/False).  
**Waarom handig?**  
Ideaal voor het filteren van lijsten op basis van een andere lijst.

### 11. **`itertools.pairwise`**
```python
from itertools import pairwise
pairwise_example([1, 2, 3, 4])
```
**Uitleg:**  
- `pairwise()` maakt paren van opeenvolgende elementen.  
**Waarom handig?**  
Ideaal voor berekeningen zoals verschillen tussen opeenvolgende waardes of afgeleiden.

### 12. **`itertools.product`**
```python
from itertools import product
product_example([1, 2], ["a", "b"])
```
**Uitleg:**  
- `product()` geeft de cartesische product van iterables.  
**Waarom handig?**  
Perfect voor combinaties bij brute-force problemen, zoals wachtwoorden genereren.

````python
from itertools import product
list(product([1, 2], ["a", "b"]))
````

### 13. **`itertools.permutations`**
```python
from itertools import permutations
permutations_example("ABCD", 2)
```
**Uitleg:**  
- `permutations()` genereert alle mogelijke volgordes van gegeven grootte.  
**Waarom handig?**  
Nuttig voor combinatorische problemen zoals puzzels of planningsproblemen.

Permutatie = Position → volgorde telt, dus ('B', 'A') ≠ ('A', 'B').

````python
from itertools import permutations
len(list(permutations("ABCD", 2)))
````

### 14. **`itertools.combinations`**
```python
from itertools import combinations
combinations_example("ABCD", 2)
```
**Uitleg:**  
- `combinations()` genereert combinaties zonder herhaling.  
**Waarom handig?**  
Veel gebruikt in statistiek, kansberekeningen en het genereren van subsets.

Combinatie = Collection → enkel de samenstelling telt.

Hier telt ('A', 'B') hetzelfde als ('B', 'A'),
dus die laatste wordt niet opnieuw opgenomen.

````python
from itertools import combinations
list(combinations("ABCD", 2))
````
