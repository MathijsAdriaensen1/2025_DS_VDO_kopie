### 1. Wat is functioneel programmeren?
Functioneel programmeren is een programmeerparadigma dat de nadruk legt op **declaratieve** code, waarbij je beschrijft wat er moet gebeuren in plaats van hoe het moet gebeuren. 

````python
# Imperatief programmeren: hoe het moet gebeuren, stap voor stap: 
# We willen de som van de kwadraten van even getallen berekenen
numbers = [1, 2, 3, 4, 5, 6]
result = 0
for n in numbers:
    if n % 2 == 0:
        result += n ** 2
print(result)
````

In imperatieve code gebruik je óók functies; het verschil is dat jij de stappen en volgorde (loops, mutaties) expliciet uitschrijft en vaak mutable state wijzigt.

````python
# Functioneel programmeren:
# Je beschrijft wat er moet gebeuren met de data, niet hoe de iteratie of accumulatie precies werkt.
# De hoe-details worden overgelaten aan de ingebouwde functies (filter, map, sum).
numbers = [1, 2, 3, 4, 5, 6]
result = sum(map(lambda n: n ** 2, filter(lambda n: n % 2 == 0, numbers)))
print(result)

# OF, nog compacter:
result = sum(n ** 2 for n in [1, 2, 3, 4, 5, 6] if n % 2 == 0)
````

Functioneel = wat + compositie. In functionele code beschrijf je het resultaat als samenstelling van (pure) functies. De iteratie/accumulatie verstop je in algemene bouwstenen (map, filter, sum, reduce) of combinators. Je vermijdt bijwerkingen en mutaties.

*Compositie* (of function composition) betekent dat je functies combineert zodat de uitvoer van de ene functie de invoer wordt van de volgende — net zoals schakels in een ketting.

````python
# Voorbeeldcode imperatief vs functioneel:
def is_even(n): return n % 2 == 0
def square(n): return n * n

# Imperatief (hoe)
nums = [1,2,3,4,5,6]
total = 0
for n in nums:
    if is_even(n):
        total += square(n)

# Functioneel (wat)
nums = [1,2,3,4,5,6]
total = sum(map(square, filter(is_even, nums)))
````

Belangrijke kenmerken zijn:
- **Pure functies**: functies zonder bijwerkingen.
- **Immutabiliteit**: data wordt niet gewijzigd, maar nieuwe data wordt gegenereerd.
- **Functies als first-class citizens**: functies kunnen worden doorgegeven als argumenten, worden opgeslagen in variabelen en worden geretourneerd door andere functies.

### 2. Pure functies
Pure functies zijn functies die:
1. Geen bijwerkingen hebben:
    - Ze wijzigen geen variabelen buiten hun scope.
    - Ze beïnvloedt bijvoorbeeld geen globale variabelen, bestanden, databases, of de console-output.
    - Ze veroorzaakt geen interactie met de "buitenwereld".
3. Altijd hetzelfde resultaat teruggeven voor dezelfde input.  We noemen dit deterministisch gedrag.  De functie is hierdoor voorspelbaar en constant.

Dus: dezelfde input → altijd dezelfde output.
Geen verborgen afhankelijkheden, geen invloed op globale variabelen of bestanden.

Hierdoor zijn pure functies makkelijker te testen.  Pure functies kunnen ook veilig parallel worden uitgevoerd omdat ze geen gedeelde variabelen aanpassen.

**Voorbeeld van een pure functie:**
```python
# Pure functie
def add(x, y):
    return x + y

print(add(2, 3))  # Output: 5
```

**Niet-pure functie:**
```python
# Niet-pure functie
z = 5

def add(x, y):
    global z
    z += 1
    return x + y + z
```

### 3. Immutability
Immutability betekent dat je data niet verandert, maar een nieuwe versie maakt als je iets wil aanpassen.

Met andere woorden: in plaats van een lijst, dict of variabele te wijzigen, maak je een nieuwe met de gewenste veranderingen.

Dit maakt code voorspelbaarder en makkelijker te debuggen.

**Voorbeeld zonder immutability (imperatief)**

Hier verander je bestaande data — dat noemen we mutable state:
```python
lst = [1, 2, 3]
lst.append(4)  # Wijzigt de originele lijst
```
De originele numbers-lijst is gewijzigd.
In een puur functionele aanpak is dat niet toegestaan, want het maakt functies moeilijker te voorspellen.

**Functioneel voorbeeld met immutability**

In een functionele stijl maak je een nieuwe lijst, zonder de oude te veranderen:
```python
lst = [1, 2, 3]
new_lst = lst + [4]  # Maakt een nieuwe lijst
```

De oorspronkelijke lijst blijft ongewijzigd.
Zo weet je altijd wat de staat van je data is!

*Waarom is dit nuttig?*

Voorspelbaar gedrag → Een functie die niets wijzigt, is makkelijk te testen.

Veiliger bij gelijktijdige uitvoering (concurrency) → Geen twee processen die dezelfde variabele aanpassen.

Geen bijwerkingen → De uitvoer hangt enkel af van de invoer.
