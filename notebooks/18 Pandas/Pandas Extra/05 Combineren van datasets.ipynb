```python
import numpy as np
from rich.console import Console
from rich.table import Table
```

```python
# transponeren van een array
array = np.array([[1, 2], [3, 4]])
print(f"Initiële array: \n{array}")
print(f"\nGetransponeerde array: \n{array.T}")
print(f"\nAlternatieve manier: \n{array.transpose()}")
```

```python
# CONCATENATIE VAN ARRAYS

array1 = np.array([[1, 2], [3, 4]])
array2 = np.array([[5, 6]])

# Concatenatie langs as 0 (rijen)
result1 = np.concatenate([array1, array2], axis=0)
print(result1, "\n")

# Concatenatie langs as 1 (kolommen)
result2 = np.concatenate([array1, array2.T], axis=1)
print(result2)
```

### 1.2 Eenvoudige concatenatie met `pd.concat`

In pandas kun je `pd.concat` gebruiken om meerdere DataFrames of Series samen te voegen. Standaard worden objecten langs de rijen (axis=0) gecombineerd.

**Voorbeeld:**

```python
import pandas as pd

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
print(f"df1:\n{df1}\n")
print(f"df2:\n{df2}\n")
# Eenvoudige concatenatie langs rijen
result = pd.concat([df1, df2])
print(f"Concatenatie langs rijen:\n{result}\n")

# Concatenatie langs kolommen
result_axis1 = pd.concat([df1, df2], axis=1)
print(f"Concatenatie langs kolommen:\n{result_axis1}\n")
```

## 2. Combining Datasets: Merge and Join

### 2.1 Basisconcept voor dataset-combinatie

Het combineren van datasets in pandas maakt gebruik van methoden die vergelijkbaar zijn met SQL-commando's zoals `JOIN`. Dit stelt je in staat om data op basis van overeenkomende kolommen te combineren.

### 2.2 Categorieën van Joins

Er zijn verschillende typen joins die kunnen worden uitgevoerd:

- **Inner join:** Alleen de rijen die overeenkomen in beide datasets worden behouden.
- **Outer join:** Alle rijen uit beide datasets worden behouden; ontbrekende waarden worden opgevuld met `NaN`.
- **Left join:** Behoudt alle rijen uit de linkertabel en voegt overeenkomende rijen uit de rechtertabel toe.
- **Right join:** Behoudt alle rijen uit de rechtertabel en voegt overeenkomende rijen uit de linkertabel toe.

**Voorbeeld:**

```python
# df1 = pd.DataFrame({'key': ['A', 'B', 'C'], 'value1': [1, 2, 3]})
# df2 = pd.DataFrame({'key': ['B', 'C', 'D'], 'value2': [4, 5, 6]})

# # Initiële dataframes
# print(f"DataFrame 1:\n\n{df1}\n\n")
# print(f"DataFrame 2: {df2}\n\n")

# # Inner join
# df_inner = pd.merge(df1, df2, on='key', how='inner')
# print(f"Merge via inner join: \n\n{df_inner}\n\n")

# # Outer join
# df_outer = pd.merge(df1, df2, on='key', how='outer')
# print(f"Merge via outer join: \n\n{df_outer}")

import pandas as pd
from rich.console import Console
from rich.table import Table

# Rich console object
console = Console()

# DataFrames definiëren
df1 = pd.DataFrame({'key': ['A', 'B', 'C'], 'value1': [1, 2, 3]})
df2 = pd.DataFrame({'key': ['B', 'C', 'D'], 'value2': [4, 5, 6]})

# Functie om een DataFrame mooi weer te geven
def print_dataframe(title, df):
    table = Table(title=title, show_header=True, header_style="bold magenta")
    
    # Voeg kolomnamen toe met kleur
    for col in df.columns:
        table.add_column(col, style="cyan", justify="center")
    
    # Voeg rijen toe
    for _, row in df.iterrows():
        table.add_row(*map(str, row))
    
    console.print(table)

# Initiale DataFrames weergeven
print_dataframe("DataFrame 1", df1)
print_dataframe("DataFrame 2", df2)

# Inner join
df_inner = pd.merge(df1, df2, on='key', how='inner')
print_dataframe("Merge via Inner Join", df_inner)

# Outer join
df_outer = pd.merge(df1, df2, on='key', how='outer')
print_dataframe("Merge via Outer Join", df_outer)
```

```python
console = Console()

def print_dataframe(title, df):
    table = Table(title=title, show_header=True, header_style="bold magenta")
    for col in df.columns:
        table.add_column(col, style="cyan")
    for _, row in df.iterrows():
        table.add_row(*map(str, row))
    console.print(table)

print_dataframe("DataFrame 1", df1)
print_dataframe("DataFrame 2", df2)
print_dataframe("Merge via Inner Join", df_inner)
print_dataframe("Merge via Outer Join", df_outer)
```

### 2.3 Specificatie van de Merge Sleutel

De `on`-parameter specificeert de kolom(len) die wordt gebruikt als sleutel voor de join.

**Voorbeeld:**

```python
# Merge met meerdere sleutels
df3 = pd.DataFrame({'key1': ['A', 'B', 'C'], 'key2': [1, 2, 3], 'value1': [10, 20, 30]})
print_dataframe("df3:", df3)
df4 = pd.DataFrame({'key1': ['A', 'B', 'D'], 'key2': [1, 2, 4], 'value2': [40, 50, 60]})
print_dataframe("\ndf4:", df4)
result = pd.merge(df3, df4, on=['key1', 'key2'], how='inner')
# print(result)
print_dataframe("\nMerge via Inner Join op 2 keys:", result)
```

### 2.4 Specificeren van set-operations voor joins

Set-operations staan voor de logische bewerkingen die worden uitgevoerd op verzamelingen (sets) van gegevens.  Bij het combineren van datasets, zoals in joins, betekent dit dat je de relaties tussen datasets kunt onderzoeken en manipuleren, zoals:

- Unie (Union): Alle unieke elementen van beide datasets.
- Doorsnede (Intersection): Alleen de elementen die in beide datasets voorkomen.
- Verschil (Difference): De elementen die uniek zijn voor een bepaalde dataset.

In de context van pandas kun je met de parameter indicator=True tijdens een join zien of een rij afkomstig is uit de linker dataset, de rechter dataset, of beide (vergelijkbaar met set-operaties). Dit biedt extra inzicht in hoe de datasets overlappen of verschillen. Gebruik dus `indicator=True` om aan te geven welke tabel de gegevens levert.

**Voorbeeld:**

```python
# even de te mergen dataframes opnieuw weergeven:
print_dataframe("DataFrame 1", df1)
print_dataframe("DataFrame 2", df2)


result = pd.merge(df1, df2, on='key', how='outer', indicator=True)
print_dataframe("Merged dataframe",result)
```

### 2.5 Overlappende kolomnamen: Het `suffixes`-sleutelwoord

Overlapping in kolomnamen kan verwarring veroorzaken bij het analyseren van gegevens.  Bij overlapping in kolomnamen kun je de `suffixes`-parameter gebruiken om duidelijk te maken uit welke dataset een specifieke kolom afkomstig is, door achtervoegsels toe te voegen aan de kolomnamen.

**Voorbeeld:**

- In de datasets df5 en df6 hebben beide een kolom genaamd value.
- Bij het samenvoegen (merge) van deze datasets zouden de value-kolommen conflicteren.
- Met suffixes=('\_left', '\_right') worden de kolommen onderscheiden als value\_left en value\_right, wat duidelijk maakt welke kolom uit welke dataset komt.

```python
df5 = pd.DataFrame({'key': ['A', 'B'], 'value': [1, 2]})
df6 = pd.DataFrame({'key': ['A', 'B'], 'value': [3, 4]})

result = pd.merge(df5, df6, on='key', suffixes=('_left', '_right'))
# print(result)
print_dataframe("Merged dataframe",result)
```

### 2.6 Voorbeeld: Data van Amerikaanse staten

We combineren datasets over de Amerikaanse staten.

1. De datasets:

- states bevat informatie over de staten en hun populatie.
- abbreviations bevat informatie over de staten en hun afkortingen.

2. Left Join:

- Combineert beide datasets op de kolom State.
- Alle rijen van states blijven behouden.
- Als een staat niet voorkomt in abbreviations, wordt NaN ingevuld voor de kolom Abbreviation.

3. Outer Join:

- Combineert beide datasets en behoudt alle rijen uit beide datasets.
- Voor staten die alleen in één van de datasets voorkomen, worden NaN-waarden toegevoegd voor ontbrekende gegevens.

4. Doel:

- Het doel is om te laten zien hoe verschillende join-methoden invloed hebben op het resultaat en hoe pandas ontbrekende gegevens (NaN) beheert.

**DataFrames:**

```python
states = pd.DataFrame({
    'State': ['California', 'Texas', 'New York'],
    'Population': [39538223, 29145505, 20201249]
})
print_dataframe("States:",states)

abbreviations = pd.DataFrame({
    'State': ['California', 'Texas', 'Florida'],
    'Abbreviation': ['CA', 'TX', 'FL']
})
print_dataframe("Abbreviations:",abbreviations)

# Left join
result = pd.merge(states, abbreviations, on='State', how='left')
print_dataframe("Merge na left join:",result)

# Outer join
result_outer = pd.merge(states, abbreviations, on='State', how='outer')
print_dataframe("Merge na outer join:",result_outer)
```
