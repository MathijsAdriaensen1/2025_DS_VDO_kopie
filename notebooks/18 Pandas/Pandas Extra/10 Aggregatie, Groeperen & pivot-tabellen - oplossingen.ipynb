### Opgave 1: Groeperen van een DataFrame op basis van schoolcode

Gebruik **Pandas** om de volgende dataset op te splitsen in groepen op basis van de kolom **school**. Controleer vervolgens het type van het resulterende `GroupBy`-object.

#### Gegeven DataFrame:
| Index | school | class | name | date_Of_Birth | age | height | weight | address |
|--------|--------|-------|----------------|--------------|----|--------|--------|--------|
| S1     | s001   | V     | Alberto Franco | 15/05/2002  | 12 | 173    | 35     | street1 |
| S2     | s002   | V     | Gino Mcneill   | 17/05/2002  | 12 | 192    | 32     | street2 |
| S3     | s003   | VI    | Ryan Parkes    | 16/02/1999  | 13 | 186    | 33     | street3 |
| S4     | s001   | VI    | Eesha Hinton   | 25/09/1998  | 13 | 167    | 30     | street1 |
| S5     | s002   | V     | Gino Mcneill   | 11/05/2002  | 14 | 151    | 31     | street2 |
| S6     | s004   | VI    | David Parkes   | 15/09/1997  | 12 | 159    | 32     | street4 |

### Vereisten:
1. **Groepeer het DataFrame op basis van de kolom `school`**.
2. **Controleer het type van het `GroupBy`-object**.
3. **Geef de gegroepeerde resultaten weer**.

```python
# Oplossing
import pandas as pd

# Gegeven DataFrame
data = {
    "school": ["s001", "s002", "s003", "s001", "s002", "s004"],
    "class": ["V", "V", "VI", "VI", "V", "VI"],
    "name": ["Alberto Franco", "Gino Mcneill", "Ryan Parkes", "Eesha Hinton", "Gino Mcneill", "David Parkes"],
    "date_Of_Birth": ["15/05/2002", "17/05/2002", "16/02/1999", "25/09/1998", "11/05/2002", "15/09/1997"],
    "age": [12, 12, 13, 13, 14, 12],
    "height": [173, 192, 186, 167, 151, 159],
    "weight": [35, 32, 33, 30, 31, 32],
    "address": ["street1", "street2", "street3", "street1", "street2", "street4"]
}

df = pd.DataFrame(data)

# Groeperen op schoolcode
grouped = df.groupby("school")

# Resultaten weergeven
print("Type van het GroupBy-object:", type(grouped))

# Toon de gegroepeerde data
for school, group in grouped:
    print(f"\nSchool: {school}")
    print(group)
```

### **Opgave 6: Groeperen van bestellingen per jaar en maand en totale aankoopwaarde berekenen**
Gebruik **Pandas** om het volgende DataFrame te groeperen op basis van het **jaar** en de **maand** van de besteldatum (`ord_date`). Bereken vervolgens het totale aankoopbedrag (`purch_amt`) per jaar en maand.

#### **Gegeven DataFrame:**
| Index | ord_no | purch_amt | ord_date   | customer_id | salesman_id |
|--------|--------|------------|------------|--------------|-------------|
| 0      | 70001  | 150.50     | 05-10-2012 | 3001         | 5002        |
| 1      | 70009  | 270.65     | 09-10-2012 | 3001         | 5005        |
| 2      | 70002  | 65.26      | 05-10-2012 | 3005         | 5001        |
| 3      | 70004  | 110.50     | 08-17-2012 | 3009         | 5003        |
| 4      | 70007  | 948.50     | 10-09-2012 | 3005         | 5002        |
| 5      | 70005  | 2400.60    | 07-27-2012 | 3001         | 5001        |
| 6      | 70008  | 5760.00    | 09-10-2012 | 3005         | 5001        |
| 7      | 70010  | 1983.43    | 10-10-2012 | 3001         | 5006        |
| 8      | 70003  | 2480.40    | 10-10-2012 | 3005         | 5003        |
| 9      | 70012  | 250.45     | 06-17-2012 | 3001         | 5002        |
| 10     | 70011  | 75.29      | 07-08-2012 | 3005         | 5007        |
| 11     | 70013  | 3045.60    | 04-25-2012 | 3005         | 5001        |

### **Vereisten:**
1. **Zet de `ord_date` om naar een datetime-formaat**.
2. **Groepeer de data op jaar en maand**.
3. **Bereken de totale `purch_amt` per jaar en maand**.
4. **Toon de resultaten in een DataFrame met de totalen per jaar en maand**.

```python
# Gegeven DataFrame
data = {
    "ord_no": [70001, 70009, 70002, 70004, 70007, 70005, 70008, 70010, 70003, 70012, 70011, 70013],
    "purch_amt": [150.50, 270.65, 65.26, 110.50, 948.50, 2400.60, 5760.00, 1983.43, 2480.40, 250.45, 75.29, 3045.60],
    "ord_date": ["05-10-2012", "09-10-2012", "05-10-2012", "08-17-2012", "10-09-2012", "07-27-2012", 
                 "09-10-2012", "10-10-2012", "10-10-2012", "06-17-2012", "07-08-2012", "04-25-2012"],
    "customer_id": [3001, 3001, 3005, 3009, 3005, 3001, 3005, 3001, 3005, 3001, 3005, 3005],
    "salesman_id": [5002, 5005, 5001, 5003, 5002, 5001, 5001, 5006, 5003, 5002, 5007, 5001]
}

df = pd.DataFrame(data)

# Stap 1: Zet de kolom 'ord_date' om naar datetime-formaat
df["ord_date"] = pd.to_datetime(df["ord_date"], format="%m-%d-%Y")

# Stap 2: Voeg aparte kolommen voor jaar en maand toe
df["year"] = df["ord_date"].dt.year
df["month"] = df["ord_date"].dt.month

# Stap 3: Groeperen op jaar en maand en totaal aankoopbedrag berekenen
df_grouped = df.groupby(["year", "month"])["purch_amt"].sum().reset_index()

# Stap 4: Print de resultaten
print(df_grouped)
```
