## 1. Inleiding tot Pandas

Pandas is een krachtige Python-bibliotheek voor data-analyse en manipulatie. Het biedt datastructuren zoals Series en DataFrames waarmee je efficiënt data kunt verwerken en analyseren.

### Wat zijn Ufuncs?

Ufuncs (universele functies) zijn NumPy-functies die elementgewijze operaties uitvoeren op arrays. In Pandas zijn deze functies uitgebreid met indexgevoeligheid, wat betekent dat de operaties rekening houden met de structuur en labels van de index.

## 2. Ufuncs: Behoud van Indexen

Wanneer je een Ufunc uitvoert op een Pandas Series of DataFrame, blijven de indexen behouden. Dit maakt het eenvoudig om de resultaten te koppelen aan de oorspronkelijke data. Hieronder een voorbeeld met een Series en een DataFrame.&#x20;

Wat is het verschil tussen een NumPy-array en een Pandas Series?&#x20;

Een NumPy-array is een homogene datastructuur die uitsluitend waarde-gebaseerde indices (op basis van volgorde) gebruikt, terwijl een Pandas Series een een-dimensionale array is met een label-gebaseerde index. Hierdoor biedt een Pandas Series meer flexibiliteit voor data-analyse omdat je toegang kunt krijgen tot waarden met behulp van expliciete labels in plaats van numerieke indices. 

### Voorbeeld 1: Elementgewijze bewerking met een Series

```python
import pandas as pd
import numpy as np

# Maak een Series
s = pd.Series([1, 2, 3, 4], index=['a', 'b', 'c', 'd'])

s
```

```python
# Pas een Ufunc toe
result = np.square(s)

print(result)
```

De index ('a', 'b', 'c', 'd') blijft behouden in het resultaat!

### Voorbeeld 2: Elementgewijze bewerking met een DataFrame

```python
# Maak een DataFrame
data = pd.DataFrame({
    'x': [1, 2, 3],
    'y': [4, 5, 6]
}, index=['a', 'b', 'c'])

# Pas een Ufunc toe
result = np.sqrt(data)

print(result)
```

De structuur en indexen blijven intact!

## 3. Ufuncs: Afstemming van Indexen

Bij operaties tussen Series of DataFrames worden indexen automatisch op elkaar afgestemd. Dit betekent dat Pandas ervoor zorgt dat de bijbehorende waarden op basis van hun labels met elkaar worden gecombineerd, zelfs als de indexen niet volledig overlappen. Dit maakt analyses intuïtiever en flexibeler. Wanneer een bepaalde waarde ontbreekt in een van de datasets, vult Pandas dit aan met NaN (Not a Number), wat helpt om ontbrekende gegevens eenvoudig te identificeren en ermee te werken.

### Voorbeeld 1: Operatie tussen twee Series

```python
# Twee Series met overlappende en unieke indexen
s1 = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
s1
```

```python
s2 = pd.Series([4, 5, 6], index=['b', 'c', 'd'])
s2
```

```python
# Operatie uitvoeren
result = s1 + s2

print(result)
```

De operatie stemt de indexen van beide Series af. Waar er geen overeenkomst is, vult Pandas dit aan met NaN.

### Voorbeeld 2: Operatie tussen twee DataFrames

```python
# Twee DataFrames met gedeeltelijk overlappende indexen en kolommen
data1 = pd.DataFrame({
    'x': [1, 2, 3],
    'y': [4, 5, 6]
}, index=['a', 'b', 'c'])

data1
```

```python
data2 = pd.DataFrame({
    'y': [7, 8, 9],
    'z': [10, 11, 12]
}, index=['b', 'c', 'd'])

data2
```

```python
# Operatie uitvoeren
result = data1 + data2

print(result)
```

De indexen en kolommen worden afgestemd en missende waarden worden aangevuld met NaN.

## 4. Ufuncs: Operaties tussen DataFrame en Series

Dit wordt ook wel broadcasting genoemd. Een operatie tussen een DataFrame en een Series wordt uitgevoerd langs de bijbehorende assen.

### Voorbeeld 1: Operatie met een kolom (verticale broadcasting)

```python
# DataFrame en Series
data = pd.DataFrame({
    'x': [1, 2, 3],
    'y': [4, 5, 6]
}, index=['a', 'b', 'c'])

data
```

```python
s = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
s
```

```python
# Operatie uitvoeren
result = data.add(s, axis=0)

print(result)
```

De operatie wordt uitgevoerd langs de rijen (indexen).

### Voorbeeld 2: Operatie met een rij (horizontale broadcasting)

```python
# DataFrame en Series
data = pd.DataFrame({
    'x': [1, 2, 3],
    'y': [4, 5, 6]
}, index=['a', 'b', 'c'])
data
```

```python
s = pd.Series({'x': 1, 'y': 2})
s
```

```python
# Operatie uitvoeren
result = data + s

print(result)
```

De operatie wordt uitgevoerd langs de kolommen.

## Samenvatting

- **Behoud van Indexen:** Ufuncs behouden de indexen en labels van Series en DataFrames.
- **Afstemming van Indexen:** Bij operaties tussen Series/DataFrames worden indexen op elkaar afgestemd, met NaN voor missende waarden.
- **Broadcasting:** Operaties tussen DataFrames en Series worden uitgevoerd langs de relevante assen.
