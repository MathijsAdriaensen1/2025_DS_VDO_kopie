### **Opgave 1: Verwijderen van spaties in een Pandas Index**
Schrijf een **Pandas-programma** om witruimtes te verwijderen uit de stringwaarden van een gegeven Pandas **Index**. Voer de volgende bewerkingen uit:
1. **Verwijder alle voor- en achterliggende spaties**.
2. **Verwijder alleen de spaties aan de linkerkant**.
3. **Verwijder alleen de spaties aan de rechterkant**.

#### **Gegeven gegevens:**
```python
[' Green', 'Black ', '  Red ', 'White', ' Pink ']
```

````python
### **Oplossing:**

import pandas as pd

# Gegeven Pandas Index met spaties
color1 = pd.Index([' Green', 'Black ', '  Red ', 'White', ' Pink '])

print("Originele index:")
print(color1)

# 1. Verwijder alle witruimtes (links en rechts)
print("\nVerwijder alle witruimtes:")
print(color1.str.strip())

# 2. Verwijder alleen de spaties aan de linkerkant
print("\nVerwijder alleen linkszijdige witruimte:")
print(color1.str.lstrip())

# 3. Verwijder alleen de spaties aan de rechterkant
print("\nVerwijder alleen rechtszijdige witruimte:")
print(color1.str.rstrip())
````

### **Opgave 2: Vind de index van een substring in een Pandas DataFrame-kolom**
Schrijf een **Pandas-programma** om de index te vinden van een opgegeven **substring** in een kolom van een DataFrame. In deze oefening gaan we zoeken naar de **positie** van de substring `"22"` in de kolom **"name_code"**.

#### **Gegeven DataFrame:**
| name_code | date_of_birth | age  |
|-----------|--------------|------|
| C001      | 12/05/2002   | 18.5 |
| c002      | 16/02/1999   | 21.2 |
| C022      | 25/09/1998   | 22.5 |
| c2002     | 12/02/2022   | 22.0 |
| C2222     | 15/09/1997   | 23.0 |

#### **Vereisten:**
1. **Zoek de indexpositie** van de substring `"22"` in de kolom **"name_code"**.
2. **Voeg een nieuwe kolom `"Index"` toe** aan het DataFrame met de gevonden indexwaarden.
3. **Toon het bijgewerkte DataFrame** met de nieuwe kolom.

````python
### **Oplossing in Pandas:**

# Gegeven DataFrame
df = pd.DataFrame({
    'name_code': ['C001', 'c002', 'C022', 'c2002', 'C2222'],
    'date_of_birth': ['12/05/2002', '16/02/1999', '25/09/1998', '12/02/2022', '15/09/1997'],
    'age': [18.5, 21.2, 22.5, 22, 23]
})

print("Origineel DataFrame:")
print(df)

# Zoek de index van de substring "22" in de kolom "name_code"
df['Index'] = df['name_code'].apply(lambda x: x.find('22'))

print("\nIndexpositie van de substring '22' de name_code kolom:")
print(df)
````

### **Opgave 3: Controleren of een kolom begint met een specifieke string in een Pandas DataFrame**  
Schrijf een **Pandas-programma** om te controleren of de waarden in een bepaalde kolom beginnen met een opgegeven string.  

#### **Gegeven DataFrame:**
| company_code | date_of_sale | sale_amount |
|-------------|-------------|------------|
| Abcd        | 12/05/2002  | 12348.5    |
| EFGF        | 16/02/1999  | 233331.2   |
| zefsalf     | 25/09/1998  | 22.5       |
| sdfslew     | 12/02/2022  | 2566552.0  |
| zekfsdf     | 15/09/1997  | 23.0       |

#### **Vereisten:**  
1. **Controleer** of elke waarde in de kolom `"company_code"` **begint met de string `"ze"`**.  
2. **Voeg een nieuwe kolom `"company_code_starts_with"`** toe die aangeeft of de waarde begint met `"ze"` (True/False).  
3. **Toon het bijgewerkte DataFrame**.  

````python
# Gegeven DataFrame
df = pd.DataFrame({
    'company_code': ['Abcd', 'EFGF', 'zefsalf', 'sdfslew', 'zekfsdf'],
    'date_of_sale': ['12/05/2002', '16/02/1999', '25/09/1998', '12/02/2022', '15/09/1997'],
    'sale_amount': [12348.5, 233331.2, 22.5, 2566552.0, 23.0]
})

print("Origineel DataFrame:")
print(df)

# Controleer of 'company_code' begint met "ze"
df['company_code_starts_with'] = df['company_code'].str.startswith('ze')

print("\nDataFrame met controle op 'ze':")
print(df)
````

### **Opgave 4: Maak een tijdreeks (time-series) in Pandas vanuit een lijst met datums**  
Schrijf een **Pandas-programma** om een **tijdreeks (time-series)** te maken op basis van een gegeven lijst van datums als strings.  

#### **Gegeven lijst met datums:**  
```python
['2014-08-01', '2014-08-02', '2014-08-03', '2014-08-04']
```
Elke datum wordt geassocieerd met een willekeurige numerieke waarde.

````python
### **Oplossing in Pandas:**  

import numpy as np

# Gegeven lijst met datums als strings
dates = ['2014-08-01', '2014-08-02', '2014-08-03', '2014-08-04']

# Maak een time-series met willekeurige waarden
time_series = pd.Series(np.random.randn(len(dates)), index=pd.to_datetime(dates))

print("Gegenereerde tijdreeks:")
print(time_series)
````
