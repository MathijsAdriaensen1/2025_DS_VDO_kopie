# Introductie tot Pandas: Aggregatie en Groeperen & pivot-tabellen

## Aggregatie en Groeperen

Het doel van aggregatie, groeperen en pivot-tabellen is om grote datasets te analyseren en samenvattingen te maken die belangrijke trends of inzichten onthullen.  

- Aggregatie wordt gebruikt voor het berekenen van samenvattende statistieken zoals gemiddelden of totalen.
- Groeperen helpt bij het opdelen van gegevens in categorieën om analyses binnen elke categorie uit te voeren.
- Pivot-tabellen bieden een krachtige manier om gegevens te herschikken en visueel inzicht te krijgen in relaties tussen verschillende dimensies.

### Planeten Data

#### Inleiding

We beginnen met een dataset over planeten die ontdekt zijn door verschillende methodes. Deze dataset bevat kolommen zoals de naam van de planeet, massa, ontdekjaar, enz. Het doel is om basisconcepten zoals aggregatie en groepering te illustreren.


```python
import pandas as pd
import seaborn as sns

# Laden van de voorbeeld dataset
data = pd.read_csv('data/planets.csv')
data.head()
```

```python
# Dataset Verkennen
# Basisinformatie over de dataset
print(data.info())

# Samenvatting van numerieke kolommen
print(data.describe())
```

### Eenvoudige Aggregaties in Pandas

#### Basisaggregaties

Hier voeren we basisaggregaties uit zoals het berekenen van het gemiddelde, de som, de minimale en maximale waarden, en het tellen van frequenties. Deze aggregaties helpen om snel inzichten te krijgen in belangrijke eigenschappen van datasets, zoals trends en uitschieters. Bijvoorbeeld, het berekenen van de gemiddelde massa van planeten geeft een idee van de typische grootte, terwijl de totale afstand inzichten biedt in de schaal van ontdekte planeten.

```python
# Gemiddelde massa van de planeten
average_mass = data['mass'].mean()
print(f"Gemiddelde massa: {average_mass}")

# Maximaal aantal ontdekkingen per jaar
max_discoveries = data['year'].value_counts().max()
print(f"Maximaal aantal ontdekkingen in een jaar: {max_discoveries}")
```

#### Handige Aggregatiefuncties

```python
# Minimum en maximum waarden
min_mass = data['mass'].min()
max_mass = data['mass'].max()

print(f"Minimum massa: {min_mass}, Maximum massa: {max_mass}")

# Totale som van afstand
total_distance = data['distance'].sum()
print(f"Totale afstand: {total_distance}")
```

### GroupBy: Split, Apply, Combine

#### Het GroupBy-Object

Het GroupBy-object in Pandas wordt gebruikt om gegevens te groeperen op basis van een of meer kolommen. Dit proces wordt vaak beschreven als "splitsen, toepassen en combineren" (split, apply, combine). Bij het splitsen worden gegevens opgesplitst in groepen, vervolgens worden functies zoals aggregatie toegepast op elke groep, en ten slotte worden de resultaten gecombineerd in een samenvattende structuur. Deze aanpak maakt het eenvoudig om analyses uit te voeren binnen specifieke categorieën of groepen in de dataset.

```python
# Groeperen op ontdekjaar
# ! groupby() retourneerteen GroupBy-object, niet een DataFrame met samengevoegde waarden
# Voeg .sum(), .mean(), .count() of een andere functie toe om de resultaten te zien.
# Gebruik .groups om te zien hoe de data is verdeeld.
grouped = data.groupby('year')

print("Groepen en bijbehorende indexen:\n")
print(grouped.groups)  # Toont de groepen en bijbehorende indexen. De lengte van de list geeft aan hoeveel planeten elk jaar zijn ontdekt

print("\n\nAantal ontdekkingen per jaar:\n")
grouped_count = data.groupby('year').count()
print(grouped_count['method'])


# Groeperen heeft pas zin als je bijvoorbeeld de som, het gemiddelde of een telling per groep wilt berekenen!
```

#### Toepassen van Aggregaties op Groepen

```python
# Gemiddelde massa per ontdekjaar
mean_mass_by_year = grouped['mass'].mean()
print(mean_mass_by_year)

# Totale afstand per methode
total_distance_by_method = data.groupby('method')['distance'].sum()
print(total_distance_by_method)
```

#### Meerdere Aggregaties

```python
# Meerdere aggregaties tegelijk
agg_results = data.groupby('method')['mass'].agg(['mean', 'max', 'count'])
print(agg_results)
```

Hier voeren we **meerdere aggregaties tegelijk** uit met behulp van de .agg()-methode. Dit stelt ons in staat om verschillende statistieken in één enkele opdracht te berekenen. In dit voorbeeld:  
- mean berekent het gemiddelde van de massa’s voor elke methode.
- max geeft de maximale massa per methode.
- count telt het aantal waarnemingen per methode.

## Pivot Tables

### Waarom Pivot-tabellen?

#### Probleemstelling

Stel je wilt de gemiddelde massa van planeten groeperen op basis van zowel het ontdekkingsjaar als de ontdekkingsmethode. Dit kan complex worden met GroupBy, maar is eenvoudig met een pivot table.

### Draaitabellen - Handmatig

#### Handmatig Groeperen

```python
# Groeperen met GroupBy
pivot_like = data.groupby(['method', 'year'])['mass'].mean().unstack()
print(pivot_like)
```

De methode `.unstack()` wordt gebruikt om een niveau van de index van een DataFrame naar kolommen te verplaatsen. Dit is nuttig wanneer je een gegroepeerde tabel wilt herstructureren naar een meer overzichtelijke vorm, vergelijkbaar met een pivot-table. Bijvoorbeeld, in het geval van de planeten data, wordt de jaarinformatie naar kolommen verplaatst om de gemiddelde massa per jaar en methode te visualiseren.

### Pivot Table Syntax

#### Basis Pivot Table

![alt text](images/pivot.png "Optionele titel")

```python
# Basis pivot table
pivot = data.pivot_table(values='mass', index='method', columns='year', aggfunc='mean')
print(pivot)
```

#### Extra Opties

```python
# Voeg marges toe voor totalen: de parameter margins=True in pivot_table() voegt totaalrijen en -kolommen toe 
# aan de pivot table. Dit betekent dat je een extra rij en kolom krijgt met de gemiddelde waarde van de volledige dataset, 
# alsof je een "Total"-rij en -kolom toevoegt.
pivot_with_totals = data.pivot_table('mass', index='method', columns='year', aggfunc='mean', margins=True)
print(pivot_with_totals)
```
