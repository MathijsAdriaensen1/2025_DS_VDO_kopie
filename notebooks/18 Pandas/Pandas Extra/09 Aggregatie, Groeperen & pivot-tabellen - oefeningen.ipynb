## Oefeningen

### Oefening 1: Aggregatie

1. Bereken de gemiddelde en maximale afstand van de planeten.
2. Tel het totaal aantal ontdekte planeten per ontdekingsmethode.

### Oefening 2: Groeperen

1. Groepeer de planeten op ontdekjaar en bereken de standaarddeviatie van hun massa.
2. Bepaal het jaar met de meeste ontdekkingen per methode.

### Oefening 3: Pivot Tables

1. Maak een pivot table van de planetendata die het aantal ontdekte planeten weergeeft per jaar en methode.
2. Voeg een totaalrij toe aan de pivot table die de gemiddelden toont.
