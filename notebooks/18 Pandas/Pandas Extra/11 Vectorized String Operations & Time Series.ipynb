### **Vectorized String Operations**

#### **1.1 Introductie tot Pandas String Operaties**

- **Doel**: Begrip krijgen van hoe Pandas efficiënt omgaat met stringbewerkingen op series.
- **Inhoud**:
  - Vectorized operations: dit zijn bewerkingen die direct worden uitgevoerd op een gehele Pandas Series of DataFrame, zonder expliciete for-loops. Dit maakt ze veel efficiënter dan traditionele iteratieve methoden.&#x20;
  - Verschillen tussen Python strings en Pandas stringmethoden:
    - Python strings zijn ontworpen voor individuele stringbewerkingen, terwijl Pandas stringmethoden ontworpen zijn om vectorized operations uit te voeren op hele Series of DataFrames.
    - Pandas stringmethoden bieden eenvoudig toegang tot functionaliteiten zoals zoeken, vervangen en pattern matching op datasets zonder expliciete for-loops.
- **Voorbeeld**:

```python
import pandas as pd

data = pd.Series(["Python", "Pandas", "Data Science", "AI"])
print(data.str.upper())  # Converteer naar hoofdletters
```

#### **1.2 Tables of Pandas String Methods**

- **Enkele beschikbare stringmethoden:**
  - `str.contains`, `str.replace`, `str.startswith`, etc.
- **Voorbeelden**:

```python
names = pd.Series(["Alice", "Bob", "Charlie", "Diana"])

filtered = names[names.str.startswith("A")]
print(f"Gefilterd dataframe met enkel namen die starten met 'A':\n{filtered}\n")

# Voorbeeld met str.contains en str.replace
contains_charlie = names[names.str.contains("Charlie")]
print(f"Gefilterd dataframe met enkel namen die 'Charlie' bevatten:\n{contains_charlie}\n")

replaced_names = names.str.replace("Bob", "Robert")
print(f"Dataframe waarbij 'Bob' is vervangen door 'Robert':\n{replaced_names}\n")
```

#### **1.3 Example: Recipe Database**

- **Doel**: Toepassen van stringmethoden op een  dataset.
- **Inhoud**:
  - Laden van een dataset met recepten.
  - Analyseren van ingrediënten of recepten met specifieke patronen.
- **Voorbeeld**:

```python
recipes = pd.Series([
    "1 cup of sugar",
    "2 cups of flour",
    "1 teaspoon of salt",
    "1/2 cup of butter"
])
contains_cup = recipes[recipes.str.contains("cup")]
print(contains_cup)
```

### **Deel 2: Werken met Time Series**

#### **2.1 Dates and Times in Python**

- **Doel**: Begrijpen hoe Python standaard datums en tijden behandelt.
- **Inhoud**:
  - `datetime` biedt standaard hulpmiddelen voor het werken met datums en tijden, zoals het maken, formatteren en manipuleren van objecten. `dateutil` breidt deze functionaliteit uit door flexibele parsing van datums mogelijk te maken en meer geavanceerde tijdsfuncties te bieden, zoals relatieve tijdsdeltas.
  - Parsing en manipulatie van datums.
- **Voorbeeld**:

```python
from datetime import datetime
from dateutil.relativedelta import relativedelta
from dateutil.parser import parse

# datetime voorbeeld - by the way: strftime staat voor string format time
date = datetime(2023, 12, 25)
print(date.strftime("%Y-%m-%d"))  # Formatteer datum als string

# dateutil relatieve tijdsdelta
date_future = date + relativedelta(months=3)
print(date_future.strftime("%Y-%m-%d"))  # Datum 3 maanden later

# dateutil flexibele parsing
parsed_date = parse("25th December 2023")
print(parsed_date.strftime("%Y-%m-%d"))  # Parse ongestructureerde datumstring
```

#### **2.2 Pandas Time Series: Indexing by Time**

- **Doel**: Het gebruik van tijd als index in Pandas is een krachtige techniek waarmee datasets kunnen worden geanalyseerd en gefilterd op basis van tijdstempels. Dit wordt gedaan om tijdreeksen overzichtelijk te organiseren, trends in tijdsgebonden gegevens te analyseren, en specifieke datumbereiken eenvoudig te extraheren..
- **Inhoud**:
  - Creëren van een tijdsindex.
  - Toegang tot data gebaseerd op tijd.

```python
import pandas as pd
# maak een pandas series waarin waarden 0 tot 9 zijn toegewezen aan een reeks datums als index,
# gegenereerd door pd.date_range
dates = pd.date_range("2023-01-01", periods=10) # standaard freq voor periods is dagen
data = pd.Series(range(10), index=dates) 
print(data["2023-01-03":"2023-01-05"])  # Data filteren op datum
```

#### **2.3 Pandas Time Series Data Structures**

- **Doel**: Tijdgebaseerde datastructuren zijn specifiek ontworpen voor het werken met datums en tijden in Pandas. Deze omvatten:
  - `DatetimeIndex`: een index van tijdstempels, gebruikt voor tijdreeksen. Praktisch voorbeeld: Je hebt dagelijkse verkoopdata en wilt deze analyseren op specifieke datums of periodes. Een `DatetimeIndex` maakt dit gemakkelijk.
  - `PeriodIndex`: een index gebaseerd op tijdsperioden zoals maanden of jaren. Praktisch voorbeeld: Je hebt maandelijkse inkomstengegevens en wilt deze groeperen en analyseren per kwartaal.
  - `TimedeltaIndex`: een index gebaseerd op tijdsverschillen. Praktisch voorbeeld: Je wilt analyseren hoeveel tijd er is verstreken tussen opeenvolgende gebeurtenissen in een logbestand of dataset.
- **Voorbeeld**:

```python
# Voorbeeld van DatetimeIndex
datetime_index = pd.date_range("2023-01-01", periods=5, freq="D")
print("DatetimeIndex:")
print(datetime_index)

# Voorbeeld van PeriodIndex
period_index = pd.period_range("2023-01", periods=5, freq="M")
print("\n\nPeriodIndex:")
print(period_index)

# Voorbeeld van TimedeltaIndex
time_delta = pd.timedelta_range(start="0 days", periods=25, freq="1h")
print("\n\nTimedeltaIndex:")
print(time_delta)
```

#### **2.4 Frequencies and Offsets**

- **Doel**: Configureren van frequenties en offsets in tijdreeksen.

- **Inhoud**:

  - Overzicht van frequenties (`D`, `H`, `M`, enz.): Dit zijn codes die bepalen hoe vaak een gebeurtenis of datapunt plaatsvindt (bijv. dagelijks, maandelijks).
    - Voorbeeld: `D` staat voor dagelijks, `H` voor elk uur, `M` voor maandelijks.
  - Gebruik van `pd.offsets`: Hiermee kun je data eenvoudig verschuiven op basis van tijdsperiodes. Voorbeelden hiervan zijn:

- `MonthEnd`: Hiermee verschuif je een datum naar het einde van de maand.
  Voorbeeld:

```python
from pandas.tseries.offsets import MonthEnd
date = pd.Timestamp("2023-01-15")
shifted_date = date + MonthEnd()
print(shifted_date)
```

- `Week`: Hiermee verschuif je een datum naar het einde van de week.
  Voorbeeld:

```python
from pandas.tseries.offsets import Week
date = pd.Timestamp("2023-01-10")
shifted_date = date + Week()
print(shifted_date)
```

- `YearEnd`: Hiermee verschuif je een datum naar het einde van het jaar.
  Voorbeeld:

```python
from pandas.tseries.offsets import YearEnd
date = pd.Timestamp("2023-10-15")
shifted_date = date + YearEnd()
print(shifted_date)
```

- **Voorbeeld van frequenties**:

```python
daily = pd.date_range("2023-01-01", periods=5, freq="D")
print(daily)
```

#### **2.5 Resampling, Shifting, and Windowing**

- **Doel**: Bewerken van tijdreeksen met technieken zoals resampling en verschuiven.

Resampling verwijst naar het proces van het wijzigen van de frequentie van een tijdreeks. Dit kan op twee manieren worden gedaan:

- **Up-sampling**: Het verhogen van de frequentie, bijvoorbeeld van maandelijkse naar dagelijkse gegevens. Hierbij moeten ontbrekende waarden worden ingevuld.

  - Voorbeeld: Je wilt dagelijkse verkoopwaarden inschatten op basis van maandelijkse gegevens.

- **Down-sampling**: Het verlagen van de frequentie, bijvoorbeeld van dagelijkse naar maandelijkse gegevens. Hierbij worden waarden samengevat, zoals door gemiddelden of totalen te berekenen.

  - Voorbeeld: Je wilt maandelijkse gemiddelde temperaturen berekenen op basis van dagelijkse waarnemingen.

- **Inhoud**:

  - Resampling (`up-sampling` en `down-sampling`).
  - Shiften van data in tijd.
  - Windowing (bijv. rollende gemiddelden).

- **Voorbeeld**:

```python
import pandas as pd

data = pd.Series(range(10), index=pd.date_range("2023-01-01", periods=10))
print(f"Initieel DataFrame:\n{data}\n\n")

print(f"DataFrame met rolling window (gemiddelde over 3 perioden):\n{data.rolling(3).mean()}")  # Rollend gemiddelde van 3 perioden
```
