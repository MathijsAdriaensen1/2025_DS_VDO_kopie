### **Oefening 1: DataFrames samenvoegen**
Schrijf een **Pandas-programma** om de twee onderstaande DataFrames langs de rijen samen te voegen en alle gegevens te behouden.

#### **Gegeven DataFrames**:

**student_data1:**
| student_id | name             | marks |
|------------|----------------|--------|
| S1         | Danniella Fenton | 200    |
| S2         | Ryder Storey    | 210    |
| S3         | Bryce Jensen    | 190    |
| S4         | Ed Bernal       | 222    |
| S5         | Kwame Morin     | 199    |

**student_data2:**
| student_id | name              | marks |
|------------|------------------|--------|
| S4         | Scarlette Fisher  | 201    |
| S5         | Carla Williamson  | 200    |
| S6         | Dante Morse       | 198    |
| S7         | Kaiser William    | 219    |
| S8         | Madeeha Preston   | 201    |

#### **Uitwerkingseisen**:
- Gebruik **Pandas** om de DataFrames samen te voegen.
- De samenvoeging moet langs de **rijen (axis=0)** gebeuren.
- De resulterende DataFrame moet **alle data behouden**.

**Tip:** Gebruik de functie `pd.concat()`.

````python
# Oplossing

import pandas as pd

# Gegeven DataFrames
student_data1 = pd.DataFrame({
    'student_id': ['S1', 'S2', 'S3', 'S4', 'S5'],
    'name': ['Danniella Fenton', 'Ryder Storey', 'Bryce Jensen', 'Ed Bernal', 'Kwame Morin'],
    'marks': [200, 210, 190, 222, 199]
})

student_data2 = pd.DataFrame({
    'student_id': ['S4', 'S5', 'S6', 'S7', 'S8'],
    'name': ['Scarlette Fisher', 'Carla Williamson', 'Dante Morse', 'Kaiser William', 'Madeeha Preston'],
    'marks': [201, 200, 198, 219, 201]
})

# Samenvoegen van de DataFrames langs de rijen
result = pd.concat([student_data1, student_data2], axis=0, ignore_index=True)

# Resultaat weergeven
print(result)
````

### **Oefening 3: Een rij toevoegen aan een DataFrame**
Schrijf een **Pandas-programma** om een lijst van dictionaries of een Series aan een bestaand DataFrame toe te voegen en de gecombineerde data weer te geven.

#### **Gegeven DataFrame**:

| index | student_id | name             | marks |
|-------|------------|----------------|--------|
| 0     | S1         | Danniella Fenton | 200    |
| 1     | S2         | Ryder Storey    | 210    |
| 2     | S3         | Bryce Jensen    | 190    |
| 3     | S4         | Ed Bernal       | 222    |
| 4     | S5         | Kwame Morin     | 199    |

#### **Toe te voegen Dictionary**:

```python
{
    "student_id": "S6",
    "name": "Scarlette Fisher",
    "marks": 205
}
```

#### **Uitwerkingseisen**:
- Voeg de dictionary als een **nieuwe rij** toe aan het DataFrame.
- Gebruik de juiste Pandas-methode om een rij dynamisch toe te voegen.
- Toon het nieuwe DataFrame inclusief de toegevoegde gegevens.

````python
### Oplossing

# Gegeven DataFrame
student_data = pd.DataFrame({
    'student_id': ['S1', 'S2', 'S3', 'S4', 'S5'],
    'name': ['Danniella Fenton', 'Ryder Storey', 'Bryce Jensen', 'Ed Bernal', 'Kwame Morin'],
    'marks': [200, 210, 190, 222, 199]
})

# Nieuwe studentgegevens
new_student = {
    'student_id': 'S6',
    'name': 'Scarlette Fisher',
    'marks': 205
}

# Oplossing: DataFrame uitbreiden met een nieuwe rij
student_data = pd.concat([student_data, pd.DataFrame([new_student])], ignore_index=True)

# Resultaat weergeven
print(student_data)
````

### **Oefening 4: DataFrames samenvoegen en mergen op een gemeenschappelijke kolom**
Schrijf een **Pandas-programma** om de twee onderstaande DataFrames **langs de rijen (axis=0)** samen te voegen en vervolgens een derde DataFrame te mergen op basis van een gemeenschappelijke kolom (`student_id`).

#### **Gegeven DataFrames**:

**student_data1:**
| index | student_id | name             | marks |
|-------|------------|----------------|--------|
| 0     | S1         | Danniella Fenton | 200    |
| 1     | S2         | Ryder Storey    | 210    |
| 2     | S3         | Bryce Jensen    | 190    |
| 3     | S4         | Ed Bernal       | 222    |
| 4     | S5         | Kwame Morin     | 199    |

**student_data2:**
| index | student_id | name              | marks |
|-------|------------|------------------|--------|
| 0     | S4         | Scarlette Fisher  | 201    |
| 1     | S5         | Carla Williamson  | 200    |
| 2     | S6         | Dante Morse       | 198    |
| 3     | S7         | Kaiser William    | 219    |
| 4     | S8         | Madeeha Preston   | 201    |

**exam_data:**
| index | student_id | exam_id |
|-------|------------|---------|
| 0     | S1         | 23      |
| 1     | S2         | 45      |
| 2     | S3         | 12      |
| 3     | S4         | 67      |
| 4     | S5         | 21      |
| 5     | S6         | 33      |
| 6     | S7         | 32      |
| 7     | S8         | 30      |
| 8     | S10        | 56      |
| 9     | S11        | 80      |
| 10    | S12        | 43      |
| 11    | S13        | 12      |

#### **Uitwerkingseisen**:
1. Voeg `student_data1` en `student_data2` samen langs de **rijen (axis=0)**.
2. Merge het samengevoegde DataFrame met `exam_data` op basis van de **gemeenschappelijke kolom `student_id`**.
3. Toon het uiteindelijke DataFrame waarin de studenten die geen overeenkomst hebben in `exam_data` behouden blijven (gebruik een **left join**).

**Tip:** Gebruik `pd.concat()` en `pd.merge()`.

````python
### Oplossing

# Gegeven DataFrames
student_data1 = pd.DataFrame({
    'student_id': ['S1', 'S2', 'S3', 'S4', 'S5'],
    'name': ['Danniella Fenton', 'Ryder Storey', 'Bryce Jensen', 'Ed Bernal', 'Kwame Morin'],
    'marks': [200, 210, 190, 222, 199]
})

student_data2 = pd.DataFrame({
    'student_id': ['S4', 'S5', 'S6', 'S7', 'S8'],
    'name': ['Scarlette Fisher', 'Carla Williamson', 'Dante Morse', 'Kaiser William', 'Madeeha Preston'],
    'marks': [201, 200, 198, 219, 201]
})

exam_data = pd.DataFrame({
    'student_id': ['S1', 'S2', 'S3', 'S4', 'S5', 'S6', 'S7', 'S8', 'S10', 'S11', 'S12', 'S13'],
    'exam_id': [23, 45, 12, 67, 21, 33, 32, 30, 56, 80, 43, 12]
})

# Stap 1: Samenvoegen van student_data1 en student_data2 langs de rijen
students_combined = pd.concat([student_data1, student_data2], ignore_index=True)

# Stap 2: Mergen met exam_data op basis van student_id
final_result = students_combined.merge(exam_data, on='student_id', how='left')

# Resultaat weergeven
print(final_result)
````

### **Oefening 5: Left Join op Meerdere Sleutels**
Schrijf een **Pandas-programma** om twee DataFrames samen te voegen met een **left join**, waarbij alleen de sleutels uit het linker DataFrame behouden blijven.

#### **Gegeven DataFrames**:

**data1:**
| index | key1 | key2 | P  | Q  |
|-------|------|------|----|----|
| 0     | K0   | K0   | P0 | Q0 |
| 1     | K0   | K1   | P1 | Q1 |
| 2     | K1   | K0   | P2 | Q2 |
| 3     | K2   | K1   | P3 | Q3 |

**data2:**
| index | key1 | key2 | R  | S  |
|-------|------|------|----|----|
| 0     | K0   | K0   | R0 | S0 |
| 1     | K1   | K0   | R1 | S1 |
| 2     | K1   | K0   | R2 | S2 |
| 3     | K2   | K0   | R3 | S3 |

#### **Uitwerkingseisen**:
1. **Merge** de DataFrames met een **left join**.
2. **Gebruik de kolommen `key1` en `key2` als joinsleutels**.
3. **Behoud alle rijen uit `data1`**, ongeacht of er een overeenkomst in `data2` is.
4. **De ontbrekende waarden in `data2` moeten als `NaN` worden weergegeven.**

````python
# Gegeven DataFrames
data1 = pd.DataFrame({
    'key1': ['K0', 'K0', 'K1', 'K2'],
    'key2': ['K0', 'K1', 'K0', 'K1'],
    'P': ['P0', 'P1', 'P2', 'P3'],
    'Q': ['Q0', 'Q1', 'Q2', 'Q3']
})

data2 = pd.DataFrame({
    'key1': ['K0', 'K1', 'K1', 'K2'],
    'key2': ['K0', 'K0', 'K0', 'K0'],
    'R': ['R0', 'R1', 'R2', 'R3'],
    'S': ['S0', 'S1', 'S2', 'S3']
})

# Left join op key1 en key2
result = pd.merge(data1, data2, on=['key1', 'key2'], how='left')

# Resultaat weergeven
print(result)
````
