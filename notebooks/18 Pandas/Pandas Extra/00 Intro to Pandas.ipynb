# Gegevensmanipulatie met Pandas

In het vorige hoofdstuk zijn we dieper ingegaan op NumPy en het `ndarray`-object, dat efficiënte opslag en manipulatie van dichtgetypte arrays in Python biedt. Hier bouwen we voort op deze kennis door gedetailleerd te kijken naar de datastructuren die worden aangeboden door de Pandas-bibliotheek. Pandas is een nieuwere bibliotheek gebouwd op NumPy en biedt een efficiënte implementatie van een `DataFrame`. `DataFrame`s zijn in wezen multidimensionale arrays met bijgevoegde rij- en kolomlabels, vaak met heterogene typen en/of ontbrekende gegevens. Naast een handige opslaginterface voor gelabelde gegevens, implementeert Pandas een aantal krachtige gegevensbewerkingsoperaties die bekend zijn bij gebruikers van zowel databasesystemen als spreadsheetprogramma's.

Zoals we zagen, biedt NumPy's `ndarray`-datastructuur essentiële functies voor het type schone, goed georganiseerde gegevens dat typisch is voor numerieke rekenkundige taken. Hoewel het deze doeleinden zeer goed dient, worden de beperkingen duidelijk wanneer we meer flexibiliteit nodig hebben (bijvoorbeeld labels aan gegevens toevoegen, werken met ontbrekende gegevens, enz.) en bij pogingen om bewerkingen uit te voeren die niet goed passen bij elementgewijze broadcasting (bijvoorbeeld groeperingen, draaitabellen, enz.), die elk een belangrijk onderdeel vormen van het analyseren van minder gestructureerde gegevens die in vele vormen beschikbaar zijn in de wereld om ons heen. Pandas, en met name de `Series` en `DataFrame` objecten, bouwen voort op de NumPy-arraystructuur en bieden efficiënte toegang tot dit soort "data cleaning" taken die een groot deel van de tijd van een datawetenschapper in beslag nemen.

In dit hoofdstuk zullen we ons richten op de mechanica van het effectief gebruiken van `Series`, `DataFrame` en gerelateerde structuren. We zullen voorbeelden gebruiken die zijn ontleend aan echte datasets waar dit relevant is, maar deze voorbeelden zijn niet noodzakelijk het hoofddoel.
